**Principal Component Analysis PCA)**

##### Introducción

**En el vasto e intrincado mundo del análisis de datos, simplificar la información compleja es clave para obtener ideas claras**. El Análisis de Componentes Principales (PCA) se destaca como una herramienta poderosa en esta tarea, ayudando a desentrañar estructuras ocultas en grandes conjuntos de datos. El PCA no es solo una técnica matemática; es una puerta de entrada a una comprensión más profunda de los datos, revelando patrones y relaciones que de otro modo podrían pasar desapercibidos. Este notebook tiene como objetivo desmitificar el PCA, guiándote a través de sus fundamentos, aplicaciones prácticas y su papel crucial en el análisis de datos. Todo el código está aquí para que lo ejecutes y entiendas.


##### Varianza, Componentes Principales e Ideas Clave

Piensa en la varianza de los datos como la diversidad de la información. Una alta varianza en una variable significa que tiene un amplio rango de valores, lo que posiblemente indica una fuente rica de información. **El análisis de componentes principales (PCA) busca direcciones (componentes principales) donde esta varianza sea máxima**. El primer componente principal captura la mayor varianza, el segundo componente principal (independiente del primero) captura la segunda mayor, y así sucesivamente. De esta forma, el PCA puede reducir efectivamente la dimensionalidad mientras conserva la mayor cantidad posible de información clave. En las siguientes secciones, profundizaremos en la aplicación práctica del PCA y cómo aprovechar su poder para descubrir tesoros ocultos en tus datos.

##### Simplificando la Complejidad

El objetivo del PCA es reducir la dimensionalidad de los datos. Cuando se trabaja con conjuntos de datos que contienen numerosas variables, el PCA ayuda a simplificarlos en un conjunto más pequeño de componentes principales. **Estos componentes son nuevas variables construidas como combinaciones lineales de las variables originales**. Pero hay un giro: se eligen de manera que capturen la mayor varianza posible, es decir, la mayor cantidad de información contenida en los datos.

**El PCA no es solo una maravilla matemática; es una herramienta práctica que resuelve problemas del mundo real. Comprender sus aplicaciones permite entender por qué es un recurso fundamental en el análisis de datos**.

##### Reducción de Dimensionalidad: Simplificando lo Complejo

El principal problema que aborda el PCA es la reducción de dimensionalidad. **En un conjunto de datos con numerosas variables, no todas contribuyen de igual manera al resultado. Algunas pueden ser redundantes o, peor aún, añadir ruido.** Los datos de alta dimensión también sufren la llamada “maldición de la dimensionalidad”, donde el volumen del espacio crece exponencialmente, haciendo que los datos se vuelvan dispersos. Esta dispersión es problemática para cualquier análisis y especialmente desafiante para los algoritmos que dependen de mediciones de distancia.

**El PCA combate estos problemas transformando las variables originales en un nuevo conjunto de variables, los componentes principales, que son menos en número pero aún capturan la esencia de los datos. Esta reducción hace que el conjunto de datos sea más fácil de explorar y visualizar, y más manejable para análisis posteriores**.

Supóngase que existe una muestra con  𝑛  individuos cada uno con  𝑝  variables ( 𝑋1 ,  𝑋2 , ...,  𝑋𝑝 ), es decir, el espacio muestral tiene  𝑝  dimensiones. PCA permite encontrar un número de factores subyacentes  (𝑧<𝑝)  que explican aproximadamente lo mismo que las  𝑝  variables originales. Donde antes se necesitaban  𝑝  valores para caracterizar a cada individuo, ahora bastan  𝑧 valores. Cada una de estas  𝑧  nuevas variables recibe el nombre de componente principal.

##### Reducción de Ruido: Filtrando lo No Deseado

Los datos suelen venir acompañados de ruido: variaciones que no son de interés pero que pueden ocultar patrones significativos. El PCA ayuda en la reducción de ruido al enfatizar las direcciones donde la varianza (la señal) es alta, lo que permite atenuar el ruido. Al reconstruir los datos utilizando solo los componentes principales más significativos, el PCA filtra eficazmente el ruido, generando datos más limpios y fáciles de interpretar.

##### Adentrándonos en las Matemáticas del PCA

Explorar las matemáticas detrás del PCA puede parecer intimidante, pero es esencial para comprenderlo a fondo. Vamos a desmitificar los conceptos clave de manera amigable, asegurándonos de que no sean solo fórmulas, sino ideas que puedas entender de forma intuitiva.

##### La Base: La Matriz de Covarianza

En el corazón del PCA se encuentra la matriz de covarianza. Es una representación matemática que muestra cómo se relaciona cada variable de tu conjunto de datos con las demás. Piénsala como una tabla que responde a la pregunta: ¿Qué ocurre con las otras variables cuando una de ellas aumenta? Valores positivos indican una relación positiva (tienden a aumentar juntas), valores negativos indican una relación negativa, y un valor cercano a cero indica que no hay relación. Esta matriz es fundamental porque contiene las pistas para comprender la estructura subyacente de tus datos.

La matriz de covarianza muestra cómo varía cada variable del conjunto de datos con respecto a las demás. Es un elemento clave en el PCA, ya que ayuda a comprender las relaciones entre las distintas variables.

formula:

$\frac{\sum_{1 = 1}^{n}(X_i-\mu_{X})(Y_i-\mu_{Y})}{n-1}$

En el contexto del Análisis de Componentes Principales (PCA), los autovectores y autovalores desempeñan un papel fundamental y tienen interpretaciones prácticas:

##### Autovectores: Las Direcciones de la Varianza

Los autovectores son donde comienza a revelarse la magia del PCA. En el contexto del PCA, cada autovector representa un componente principal. **Imagina tus datos multidimensionales como una nube de puntos. Los autovectores son direcciones que atraviesan esa nube**. Por ejemplo, en un espacio tridimensional, un autovector podría ser una línea que atraviesa la nube en cualquier dirección. Lo fascinante de estos autovectores es que están orientados en las direcciones donde los datos presentan mayor variación. El primer autovector (o primer componente principal) apunta hacia donde la varianza es mayor. El segundo autovector es perpendicular al primero y señala la siguiente dirección con mayor varianza, y así sucesivamente.

**Autovectores (Componentes Principales): Representan las direcciones en el conjunto de datos donde existe mayor varianza, es decir, donde los datos están más dispersos. En PCA, se busca identificar las direcciones que mejor explican la variabilidad de los datos**. Estas direcciones son los componentes principales del conjunto de datos y corresponden a los autovectores de la matriz de covarianza. Cada componente principal capta diferentes características de la estructura de los datos.

##### Autovalores: Cuantificando la Importancia

**Cada autovector tiene un autovalor correspondiente. En el PCA, un autovalor cuantifica la cantidad de varianza capturada en la dirección de su autovector. Un autovalor más alto significa más varianza (información) en esa dirección. Por eso, a menudo se ordenan los componentes principales en orden descendente según sus autovalores: para clasificarlos según su importancia.** Esto nos indica cuán “importante” es cada dimensión para describir la varianza de los datos.

##### Paso a Paso: De la Teoría a la Práctica

- Estandarizar los Datos: El PCA comienza con la estandarización, convirtiendo las variables a una escala común.
- Calcular la Matriz de Covarianza: Esta matriz muestra cómo se relacionan entre sí las variables.
- Encontrar Autovectores y Autovalores: Se calculan a partir de la matriz de covarianza. Los autovectores nos indican los componentes principales, y los autovalores nos dicen cuán importantes son.
- Transformar los Datos: Finalmente, se transforman los datos al nuevo espacio definido por los componentes principales.

- Autovectores y Dispersión de los Datos: Imagina un gráfico en 3D de tus datos con líneas (autovectores) que lo atraviesan, indicando - - las direcciones de máxima dispersión.

**Álgebra lineal**

En esta sección se describen dos de los conceptos matemáticos que se aplican en el PCA: eigenvectors y eigenvalues. Se trata simplemente de una descripción intuitiva con la única finalidad de facilitar el entendimiento del cálculo de componentes principales.

$
\begin{pmatrix}
2 & 3\\
2 & 1
\end{pmatrix}
\begin{pmatrix}
3\\
2
\end{pmatrix}=
\begin{pmatrix}
12\\
8
\end{pmatrix}=
4\begin{pmatrix}
3\\
2
\end{pmatrix}
$

El vector resultante de la multiplicación es un múltiplo entero del vector original. Los eigenvectors de una matriz son todos aquellos vectores que, al multiplicarlos por dicha matriz, resultan en el mismo vector o en un múltiplo entero del mismo. Los eigenvectors tienen una serie de propiedades matemáticas específicas:

- Los eigenvectors solo existen para matrices cuadradas y no para todas. En el caso de que una matriz n x n tenga eigenvectors, el número de ellos es n.

- Si se escala un eigenvector antes de multiplicarlo por la matriz, se obtiene un múltiplo del mismo eigenvector. Esto se debe a que, si se escala un vector multiplicándolo por cierta cantidad, lo único que se consigue es cambiar su longitud pero la dirección es la misma.

- Todos los eigenvectors de una matriz son perpendiculares (ortogonales) entre ellos, independientemente de las dimensiones que tengan.

Dada la propiedad de que multiplicar un eigenvector solo cambia su longitud pero no su naturaleza de eigenvector, es frecuente escalarlos de tal forma que su longitud sea 1. De este modo se consigue que todos ellos estén estandarizados. A continuación se muestra un ejemplo:

El eigenvector $\begin{pmatrix}3\\2\end{pmatrix}$ tiene una lonfitud  $\sqrt{3^2+2^2}=\sqrt{13}$ 

$\begin{pmatrix}3\\2\end{pmatrix}$÷$\sqrt{13}=\begin{pmatrix}3/\sqrt{13}\\2/\sqrt{13}\end{pmatrix}$

**Eigenvalue**

Cuando se multiplica una matriz por alguno de sus eigenvectors se obtiene un múltiplo del vector original, es decir, el resultado es ese mismo vector multiplicado por un número. Al valor por el que se multiplica el eigenvector resultante se le conoce como eigenvalue. A todo eigenvector le corresponde un eigenvalue y viceversa.

En el método PCA, cada una de las componentes se corresponde con un eigenvector, y el orden de componente se establece por orden decreciente de eigenvalue. Así pues, la primera componente es el eigenvector con el eigenvalue más alto.


**Interpretación geométrica de las componentes principales**


Una forma intuitiva de entender el proceso de PCA es interpretar las componentes principales desde un punto de vista geométrico. Supóngase un conjunto de observaciones para las que se dispone de dos variables ( 𝑋1 ,  𝑋2 ). El vector que define la primera componente principal ( 𝑍1 ) sigue la dirección en la que las observaciones tienen más varianza (línea roja). La proyección de cada observación sobre esa dirección equivale al valor de la primera componente para dicha observación (principal component score,  𝑧𝑖1 ).

La segunda componente ( 𝑍2 ) sigue la segunda dirección en la que los datos muestran mayor varianza y que no está correlacionada con la primera componente. La condición de no correlación entre componentes principales equivale a decir que sus direcciones son perpendiculares/ortogonales.

Si tus datos están muy alineados en diagonal, proyectarlos sobre el eje diagonal resume casi toda la información en una sola dimensión.

Cada observación, que antes tenía coordenadas en las variables originales, ahora tiene coordenadas nuevas en el espacio de los componentes principales (PC1, PC2, …). Estas nuevas coordenadas son las proyecciones.Las proyecciones son las nuevas coordenadas de tus datos en el sistema de ejes que mejor explica su variabilidad.

Cada componente principal ( 𝑍𝑖 ) se obtiene por combinación lineal de las variables originales. Se pueden entender como nuevas variables obtenidas al combinar de una determinada forma las variables originales. La primera componente principal de un grupo de variables ( 𝑋1 ,  𝑋2 , ...,  𝑋𝑝 ) es la combinación lineal normalizada de dichas variables que tiene mayor varianza:

$𝑍_1=\phi_{11}𝑋_1+\phi_{21}𝑋_2+...+\phi_{p1}𝑋_p$

Que la combinación lineal sea normalizada implica que:

$\sum_{j = 1}^{p}\phi_{j1}^{2}=1$

Los términos  $\phi_{11} , ...,  \phi_{1p}$  reciben en el nombre de loadings y son los que definen las componentes. Por ejemplo,  $\phi_{11}$  es el loading de la variable  $X_1$  de la primera componente principal. Los loadings pueden interpretarse como el peso/importancia que tiene cada variable en cada componente y, por lo tanto, ayudan a conocer que tipo de información recoge cada una de las componentes.

Dado un set de datos  𝑋  con n observaciones y p variables, el proceso a seguir para calcular la primera componente principal es:

Centrar las variables: se resta a cada valor la media de la variable a la que pertenece. Con esto se consigue que todas las variables tengan media cero.

Se resuelve un problema de optimización para encontrar el valor de los loadings con los que se maximiza la varianza. Una forma de resolver esta optimización es mediante el cálculo de eigenvector-eigenvalue de la matriz de covarianzas.

Una vez calculada la primera componente ( $Z_1$ ), se calcula la segunda ( $Z_2$ ) repitiendo el mismo proceso pero añadiendo la condición de que la combinación lineal no pude estar correlacionada con la primera componente. Esto equivale a decir que  $Z_1$  y $Z_2$  tienen que ser perpendiculares. EL proceso se repite de forma iterativa hasta calcular todas las posibles componentes (min(n-1, p)) o hasta que se decida detener el proceso. El orden de importancia de las componentes viene dado por la magnitud del eigenvalue asociado a cada eigenvector.


**Escalado de las variables**


El proceso de PCA identifica las direcciones con mayor varianza. Como la varianza de una variable se mide en sus mismas unidades elevadas al cuadrado, si antes de calcular las componentes no se estandarizan todas las variables para que tengan media cero y desviación estándar de uno, aquellas variables cuya escala sea mayor dominarán al resto. De ahí que sea recomendable estandarizar siempre los datos.


**Reproducibilidad de las componentes**


El proceso de PCA estándar es determinista, genera siempre las mismas componentes principales, es decir, el valor de los loadings resultantes es el mismo. La única diferencia que puede darse es que el signo de todos los loadings esté invertido. Esto es así porque el vector de loadings determina la dirección de la componente, y dicha dirección es la misma independientemente del signo (la componente sigue una línea que se extiende en ambas direcciones). Del mismo modo, el valor específico de las componentes obtenido para cada observación (principal component scores) es siempre el mismo, a excepción del signo.


**Proporción de varianza explicada**


Una de las preguntas más frecuentes que surge tras realizar un PCA es: ¿Cuánta información presente en el set de datos original se pierde al proyectar las observaciones en un espacio de menor dimensión? o lo que es lo mismo ¿Cuanta información es capaz de capturar cada una de las componentes principales obtenidas? Para contestar a estas preguntas se recurre a la proporción de varianza explicada por cada componente principal.

Asumiendo que las variables se han normalizado para tener media cero, la varianza total presente en el set de datos se define como

$\sum_{j = 1}^{p}Var(X_j)=\sum_{j = 1}^{p}\frac{1}{n}\sum_{1 = 1}^{n}x_{ij}^{2}$

y la varianza explicada por la componente m es

$\frac{1}{n}\sum_{1 = 1}^{n}z_{im}^{2}=\frac{1}{n}\sum_{i = 1}^{n}(\sum_{j = 1}^{p}\phi_{jm}x_{ij})^2$

Por lo tanto, la proporción de varianza explicada por la componente m viene dada por el ratio

$\frac{\sum_{1 = 1}^{n}(\sum_{j = 1}^{p}\phi_{jm}x_{ij})^2}{\sum_{1 = 1}^{p}\sum_{1 = 1}^{n}x_{ij}^{2}}$

Tanto la proporción de varianza explicada, como la proporción de varianza explicada acumulada, son dos valores de gran utilidad a la hora de decidir el número de componentes principales a utilizar en los análisis posteriores. Si se calculan todas las componentes principales de un set de datos, entonces, aunque transformada, se está almacenando toda la información presente en los datos originales. El sumatorio de la proporción de varianza explicada acumulada de todas las componentes es siempre 1.


**Número óptimo de componentes principales**

Por lo general, dada una matriz de datos de dimensiones n x p, el número de componentes principales que se pueden calcular es como máximo de n-1 o p (el menor de los dos valores es el limitante). Sin embargo, siendo el objetivo del PCA reducir la dimensionalidad, suelen ser de interés utilizar el número mínimo de componentes que resultan suficientes para explicar los datos. No existe una respuesta o método único que permita identificar cual es el número óptimo de componentes principales a utilizar. Una forma de proceder muy extendida consiste en evaluar la proporción de varianza explicada acumulada y seleccionar el número de componentes mínimo a partir del cual el incremento deja de ser sustancial.

#### Ejercicio 1. Simulacion
4 observaciones × 2 variables.

````python
import numpy as np
import pandas as pd
from sklearn.preprocessing import scale
from sklearn.decomposition import PCA

# Datos de ejemplo: 4 observaciones y 2 variables
datos = np.array([
    [2, 8],
    [3, 6],
    [4, 5],
    [5, 3]
])

print("Datos originales:\n", datos)
print("Shape:", datos.shape)  # (4 filas, 2 columnas)
````

````python
media = np.mean(datos)
desviacion = np.std(datos)
varianza = np.var(datos)
print("Media:", media)
print("Desviación estándar:", desviacion)
print("Varianza:", varianza)
````

##### Paso 2: Escalar los datos

Para que cada variable tiene media 0 y desviación 1.

````python
datos_escalados = scale(datos)
print("\nDatos escalados:\n", datos_escalados)
print("Media por columna:", datos_escalados.mean(axis=0).round(5))
print("Desviación estándar por columna:", datos_escalados.std(axis=0).round(5))
````

##### Paso 3: Ajustar PCA

````python
modelo_pca = PCA(n_components=2)
modelo_pca.fit(datos_escalados)

print("\nComponentes principales (vectores propios):\n", modelo_pca.components_)
print("Shape:", modelo_pca.components_.shape)
````

Son los nuevos ejes estos dos componentes

##### Paso 4. Producto matricial manual

````python
proyecciones = np.dot(modelo_pca.components_, datos_escalados.T)
                       #(2x2)                    #(2x4)
print("\nProyecciones (manual con np.dot):\n", proyecciones)
print("Shape:", proyecciones.shape)
````

Forma (2, 4):

- 2 componentes
- 4 observaciones

Cada columna representa la posición de una observación en el espacio de los componentes principales.

````python
print(proyecciones.T)
print("Shape:", proyecciones.T.shape)
````

##### Paso 5: Comparación con transform

````python
proyecciones_transform = modelo_pca.transform(datos_escalados).T

print("\nProyecciones con PCA.transform (transpuesta para comparar):\n", proyecciones_transform)
````

Si no se estandarizan las variables para que tengan media cero y desviación estándar de uno antes de realizar el estudio PCA, la variable Assault, que tiene una media y dispersión muy superior al resto, dominará la mayoría de las componentes principales.

**Modelo PCA**

La clase sklearn.decomposition.PCA incorpora las principales funcionalidades que se necesitan a la hora de trabajar con modelos PCA. El argumento n_components determina el número de componentes calculados. Si se indica None, se calculan todas las posibles (min(filas, columnas) - 1).

Por defecto, PCA() centra los valores pero no los escala. Esto es importante ya que, si las variables tienen distinta dispersión, como en este caso, es necesario escalarlas. Una forma de hacerlo es combinar un StandardScaler() y un PCA() dentro de un pipeline. Para más información sobre el uso de pipelines consultar Pipeline y ColumnTransformer.

````python
# Entrenamiento modelo PCA con escalado de los datos
# ==============================================================================
pca_pipe = make_pipeline(StandardScaler(), PCA())
pca_pipe.fit(datos)

# Se extrae el modelo entrenado del pipeline
modelo_pca = pca_pipe.named_steps['pca']
````

**Interpretación**


Una vez entrenado el objeto PCA, pude accederse a toda la información de las componentes creadas.

components_ contiene el valor de los loadings  𝜙  que definen cada componente (eigenvector). Las filas se corresponden con las componentes principals (ordenadas de mayor a menor varianza explicada). Las filas se corresponden con las variables de entrada.

````python
# Se convierte el array a dataframe para añadir nombres a los ejes.
componentes = pd.DataFrame(
    data    = modelo_pca.components_,
    columns = datos.columns,
    index   = ['PC1', 'PC2', 'PC3', 'PC4']
)

componentes  # así también lo muestra en el output
````

##### Demostrando ortogonalidad 

````python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import scale

# Supongamos que tienes un DataFrame "datos" con 4 variables
datos_escalados = scale(datos)

modelo_pca = PCA(n_components=4)
modelo_pca.fit(datos_escalados)

# Scores (proyecciones de los datos en los PCs)
scores = modelo_pca.transform(datos_escalados)

scores_df = pd.DataFrame(
    data=scores,
    columns=['PC1', 'PC2', 'PC3', 'PC4']
)

# Matriz de correlaciones entre los PCs 
corr_matrix = scores_df.corr().round(1).abs()
print(corr_matrix)
````

$𝑍_1=\phi_{11}𝑋_1+\phi_{21}𝑋_2+...+\phi_{p1}𝑋_p$

Que la combinación lineal sea normalizada implica que:

$\sum_{j = 1}^{p}\phi_{j1}^{2}=1$

$0.535899^2+0.583184^2+0.278191^2+0.543432^2=1$

Cada componente principal ( 𝑍𝑖 ) se obtiene por combinación lineal de las variables originales. Se pueden entender como nuevas variables obtenidas al combinar de una determinada forma las variables originales. La primera componente principal de un grupo de variables ( 𝑋1 ,  𝑋2 , ...,  𝑋𝑝 ) es la combinación lineal normalizada de dichas variables que tiene mayor varianza.

Analizar con detalle el vector de loadings que forma cada componente puede ayudar a interpretar qué tipo de información recoge cada una de ellas. Por ejemplo, la primera componente es el resultado de la siguiente combinación lineal de las variables originales:

PC1=0.535899 Murder+0.583184 Assault+0.278191 UrbanPop+0.543432 Rape

Los pesos asignados en la primera componente a las variables Assault, Murder y Rape son aproximadamente iguales entre ellos y superiores al asignado a UrbanPoP. Esto significa que la primera componente recoge mayoritariamente la información correspondiente a los delitos. En la segunda componente, es la variable UrbanPoP es la que tiene con diferencia mayor peso, por lo que se corresponde principalmente con el nivel de urbanización del estado. Si bien en este ejemplo la interpretación de las componentes es bastante clara, no en todos los casos ocurre lo mismo, sobre todo a medida que aumenta el número de variables.

La influencia de las variables en cada componente analizarse visualmente con un gráfico de tipo heatmap.

````python
# Heatmap componentes
# ==============================================================================
fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(7, 5))
componentes = modelo_pca.components_
plt.imshow(componentes.T, cmap='viridis', aspect='auto')
plt.yticks(range(len(datos.columns)), datos.columns)
plt.xticks(range(len(datos.columns)), np.arange(modelo_pca.n_components_) + 1)
plt.grid(False)
plt.colorbar();
````

Una vez calculadas las componentes principales, se puede conocer la varianza explicada por cada una de ellas, la proporción respecto al total y la proporción de varianza acumulada. Esta información está almacenada en los atributos explained_variance_ y explained_variance_ratio_ del modelo.

````python
# Assuming pca is already fitted PCA object from sklearn
eigenvalues = pca.explained_variance_

# Applying Kaiser Criterion
kaiser_criterion = eigenvalues > 1

# Get the number of components that satisfy the Kaiser Criterion
n_components_kaiser = sum(kaiser_criterion)

# Print the components that satisfy the criterion
print(f"Number of components with eigenvalue > 1: {n_components_kaiser}")
important_components = np.arange(1, n_components_kaiser + 1)
print(f"Important components according to Kaiser Criterion: {important_components}")

# Optionally, if you want to see which specific components satisfy the criterion
for i, (ev, satisfy) in enumerate(zip(eigenvalues, kaiser_criterion), start=1):
    if satisfy:
        print(f"Component {i} with eigenvalue: {ev}")
````

### Proyecciones en PCA: utilidad e interpretación
```

Las proyecciones son las coordenadas de cada observación en el nuevo espacio de componentes principales. En tu ejemplo, cada estado de EE.UU. pasa de tener 4 variables originales (crimen, asalto, etc.) a tener 4 coordenadas en el espacio PCA.


Trasformación

Que una vez que entrenas el modelo PCA con un conjunto de datos, puedes usar .transform() para reducir esos datos a un espacio de menor dimensión, conservando la mayor parte posible de la información (varianza).
Esto es útil para:

- Visualización (por ejemplo, usando PC1 y PC2 en un gráfico 2D).
- Reducción de complejidad en modelos de ML.
- Eliminar ruido o redundancia en los datos.

````python
# Proyección de las observaciones de entrenamiento
# ==============================================================================
proyecciones = pca_pipe.transform(X=datos)
proyecciones = pd.DataFrame(
    proyecciones,
    columns = ['PC1', 'PC2', 'PC3', 'PC4'],
    index   = datos.index
)
proyecciones.head(20)
````

Tomando Alabama (0.985, -1.133, -0.444, -0.156):

- PC1 = 0.985 → Alabama está ligeramente por encima del promedio en la dirección de mayor varianza
- PC2 = -1.133 → Alabama está por debajo del promedio en la segunda dirección de varianza
  
Los valores son distancias estandarizadas desde el origen (centroide de los datos)

### Utilidades concretas
1. Reducción de dimensionalidad
Si PC1 + PC2 explican el 85% de la varianza, puedes quedarte solo con esas dos columnas y descartar PC3 y PC4 sin perder casi información.
2. Visualización (la más poderosa)
Con PC1 vs PC2 puedes graficar 50 estados en 2D y ver clusters naturales — algo imposible con 4 variables simultáneas.
3. Detección de outliers
Alaska tiene PC3 = 2.04, muy alejado de cero → es atípico en esa dimensión. Las proyecciones extremas señalan observaciones inusuales.
4. Input para modelos posteriores
Las proyecciones son features decorrelacionadas — ideales para alimentar regresión, clustering (K-Means) o clasificación, evitando multicolinealidad.
5. Comparación entre observaciones
California (PC1=2.52) vs Arkansas (PC1=-0.14): California está mucho más en el extremo positivo de la dimensión de mayor varianza.

#### Regla práctica de interpretación

Valor de proyección

Significado
- Alto positivo: Estado con alta presencia del "perfil" que define ese PC
- Cercano a cero: Estado promedio en esa dimensión
- Alto negativo: Estado con perfil opuesto al que define ese PC

La transformación es el resultado de multiplicar los vectores que definen cada componente con el valor de las variables. Puede calcularse de forma manual:

Puedo calcular las proyecciones multiplicando la matriz de PCA (4x4) por la transpuesta de la matriz X de datos escalada (4x50) dando como resulado una matriz de dimension 4x50 (que es la misma dimension de los datos originales. Ese np.dot(modelo_pca.components_, scale(datos).T) es simplemente otra forma de calcular lo que hace modelo_pca.transform(datos) en scikit-learn.

````python
proyecciones = np.dot(modelo_pca.components_, scale(datos).T)
#                             (4x4)               (4X50)
proyecciones = pd.DataFrame(proyecciones, index = ['PC1', 'PC2', 'PC3', 'PC4'])
proyecciones = proyecciones.transpose().set_index(datos.index)
proyecciones.head()
````

### Principal Components Regression


El método Principal Components Regression PCR consiste en ajustar un modelo de regresión lineal por mínimos cuadrados empleando como predictores las componentes generadas a partir de un Principal Component Analysis (PCA). De esta forma, con un número reducido de componentes se puede explicar la mayor parte de la varianza de los datos.

En los estudios observacionales, es frecuente disponer de un número elevado de variables que se pueden emplear como predictores, sin embargo, esto no implica necesariamente que se disponga de mucha información. Si las variables están correlacionadas entre ellas, la información que aportan es redundante y además, se incumple la condición de no colinealidad necesaria en la regresión por mínimos cuadrados. Dado que el PCA es útil eliminando información redundante, si se emplean como predictores las componentes principales, se puede mejorar el modelo de regresión. Es importante tener en cuenta que, si bien el Principal Components Regression reduce el número de predictores del modelo, no se puede considerar como un método de selección de variables ya que todas ellas se necesitan para el cálculo de las componentes. La identificación del número óptimo de componentes principales que se emplean como predictores en PCR puede identificarse por validación cruzada.

**PCR**

Para combinar PCA con regresión lineal, se crea un pipeline que combine ambos procesos. Dado que no se puede conocer a priori el número de componentes óptimo, se recurre a validación cruzada.

````python
# Entrenamiento modelo de regresión precedido por PCA con escalado
# ==============================================================================
pipe_modelado = make_pipeline(StandardScaler(), PCA(n_components=20), LinearRegression())
pipe_modelado.fit(X=X_train, y=y_train)
````

Este código crea y entrena un pipeline de modelado en scikit-learn que realiza tres pasos secuenciales sobre los datos de entrenamiento:

Paso a paso:

- StandardScaler()
- Escala las variables predictoras (X_train) para que tengan media 0 y desviación estándar 1. Esto es importante para que los componentes principales en PCA no estén dominados por variables con escalas mayores.
PCA()
- Aplica Análisis de Componentes Principales (PCA) para reducir la dimensionalidad de los datos transformados por StandardScaler.
- Por defecto, PCA() calcula todas las componentes posibles, pero se pueden especificar cuántas mantener con n_components.
- LinearRegression()
- Entrena un modelo de regresión lineal sobre los datos transformados por PCA.

Qué hace .fit(X=X_train, y=y_train)?
Entrena el pipeline completo:

- Primero escala X_train con StandardScaler.
- Luego aplica PCA sobre los datos escalados.
- Finalmente, ajusta el modelo LinearRegression a los componentes principales obtenidos.

¿Para qué sirve usar un pipeline?
- Organización: Encadena pasos de preprocesamiento y modelado de forma clara.
- Reproducibilidad: Asegura que exactamente los mismos pasos se apliquen al conjunto de entrenamiento y después al de prueba o nuevas observaciones.
- Evita fugas de datos: Los pasos se ajustan sólo con datos de entrenamiento
- Facilita la validación cruzada y la búsqueda de hiperparámetros
Puedes usar GridSearchCV o cross_val_score directamente sobre el pipeline, sin preocuparte por aplicar manualmente los pasos previos dentro de cada pliegue.



Primero se evalua el modelo si se incluyen todas las componentes.

````python
# Predicciones test
# ==============================================================================
predicciones = pipe_modelado.predict(X=X_test)
predicciones = predicciones.flatten()

# Error de test del modelo 
# ==============================================================================
mse_pcr = mean_squared_error(
            y_true  = y_test,
            y_pred  = predicciones)
rmse_pcr= np.sqrt(mse_pcr)
print("")
print(f"El error (rmse) de test es: {rmse_pcr}")
````

````python
from sklearn.metrics import mean_squared_error
import numpy as np

# Valores de n_components a evaluar
param_values = [1, 2, 4, 6, 8, 10, 15, 20, 30, 50]

# Guardar resultados
resultados = []

for n in param_values:
    # Ajustar el pipeline con el valor actual de n_components
    pipe_modelado.set_params(pca__n_components=n)
    pipe_modelado.fit(X_train, y_train)
    
    # Predicciones
    y_train_pred = pipe_modelado.predict(X_train)
    y_test_pred  = pipe_modelado.predict(X_test)
    
    # Calcular RMSE (positivo)
    rmse_train = np.sqrt(mean_squared_error(y_train, y_train_pred))
    rmse_test  = np.sqrt(mean_squared_error(y_test, y_test_pred))
    
    resultados.append({
        "pca__n_components": n,
        "rmse_train": rmse_train,
        "rmse_test": rmse_test
    })

# Convertir a DataFrame
resultados = pd.DataFrame(resultados)

# Ordenar por RMSE de test (menor es mejor)
resultados = resultados.sort_values("rmse_test", ascending=True)
resultados.head(10)
````

````python
import matplotlib.pyplot as plt

# Aseguramos que los componentes estén ordenados
resultados = resultados.sort_values("pca__n_components")

# Gráfico tipo serie de tiempo (line plot)
fig, ax = plt.subplots(figsize=(8, 4))

ax.plot(resultados["pca__n_components"], resultados["rmse_train"], marker="o", linestyle="-", label="Train RMSE")
ax.plot(resultados["pca__n_components"], resultados["rmse_test"], marker="o", linestyle="-", label="Test RMSE")

ax.set_title("Evolución del RMSE según número de componentes PCA")
ax.set_xlabel("Número de componentes PCA")
ax.set_ylabel("RMSE")
ax.legend()
ax.grid(True)

plt.show()
````

````python
# Mejores hiperparámetros sin validación cruzada
# ==============================================================================
print("----------------------------------------")
print("Mejores hiperparámetros encontrados (sin CV)")
print("----------------------------------------")

mejor = resultados.loc[resultados["rmse_test"].idxmin()]  # fila con menor RMSE en test

print(f"pca__n_components: {mejor['pca__n_components']}")
print(f"Best RMSE (Test): {mejor['rmse_test']:.4f}")
print(f"RMSE (Train): {mejor['rmse_train']:.4f}")
````

##### Nota: 

La razón de que en scikit-learn los errores (MSE, RMSE, MAE, etc.) aparezcan como negativos en GridSearchCV es por un tema de consistencia interna en cómo se manejan las métricas de scoring.
Si usas accuracy, r2 o f1, es lógico: más alto = mejor. Pero si usas métricas de error (ej. RMSE, MAE, MSE), ahí el criterio correcto sería más bajo = mejor.
Esto generaría una inconsistencia: a veces habría que maximizar y a veces minimizar.

Para evitar esa ambigüedad, el equipo de scikit-learn decidió que:
Todas las métricas de error se devuelven como valores negativos.
De esta forma, siempre se aplica el mismo criterio:
GridSearchCV maximiza el score (es decir, busca el valor menos negativo → error más pequeño).

#### Selecciónando cargas (loadings)

Usamos percentil 75 como umbral para seleccionar cargas importantes en PCA, no es un valor “fijo” obligatorio; es más bien una convención práctica para quedarse con las variables más influyentes. 

##### Qué significa elegir percentil 75

- Calculas el percentil 75 de todas las cargas absolutas.
- Esto selecciona solo las cargas más altas: el 25% superior.
- Variables con carga mayor que este umbral se consideran importantes para definir el componente.

##### Por qué se usa 75%
- Es un compromiso entre incluir suficiente información y no saturar cada componente.
- Si se escoge un percentil más bajo (50 o 60), se seleccionan muchas variables, incluyendo algunas que solo contribuyen débilmente.
- Si se escoge  un percentil más alto (80 o 90), se seleccionan las variables más dominantes, pero podrían perderse patrones interesantes.

##### Reglas de decisión para elegir el umbral

##### Dependiendo del objetivo del análisis:

##### Objetivo	Umbral sugerido
- Interpretar componentes y nombrarlos claramente	Percentil 75 o 80
- Analizar patrones generales, sin perder variables débiles	Percentil 50–60
- Muy estricto, solo variables dominantes	Percentil 85–90

````python
import numpy as np

# Calcular umbral como percentil 75 de las cargas absolutas
threshold = np.percentile(loadings.abs().values, 75)
print("Umbral automático:", threshold)
````

- loadings.abs().values → toma los valores absolutos de las cargas factoriales (para que no importe si son positivos o negativos).
- np.percentile(..., 75) → obtiene el percentil 75 de todos esos valores. Es decir, el valor por encima del cual están el 25% más alto de las cargas.
- threshold → será ese valor que se toma como umbral automático.
Solo se consideran importantes las variables cuyas cargas sean mayores al percentil 75, es decir, las más influyentes en cada componente.

#### Explicacion percentil 75

Piensa en una fila de estudiantes con sus notas de 0 a 100. Supongamos que tienes estas calificaciones:

$[40, 55, 60, 70, 72, 75, 78, 80, 85, 90, 95]$

Si ordenas los datos (ya lo están), el percentil 75 es la nota que marca el límite superior del 75% de los estudiantes con notas más bajas, y por encima quedan solo los mejores 25%

El percentil 75 significa: “El valor que está en la posición 75% de la distribución ordenada”.

1. Calcular la posición
- La fórmula para la posición (índice) es:
- $pos=(p/100)×(n+1)$
donde:

- p=75 (percentil 75)
  
- n=11 (número de datos)
$pos=0.75×(11+1)=0.75×12=9$

- El percentil 75 está en la posición 9 de la lista ordenada.

2. 2. Localizar el valor
Lista ordenada con posiciones:


- 1. 40
- 2. 55
- 3. 60
- 4. 70
- 5. 72
- 6. 75
- 7. 78
- 8. 80
- 9. 85 ⟵ aquí está la posición 9
- 10. 90
- 11. 95


Por lo tanto, el percentil 75 = 85. 

#### Intepolacion lineal de numpy

A continuacion esta la explicacion de fórmula de interpolación lineal de numpy.percentile (que en realidad implementa el método linear interpolation between closest ranks).

Fórmula general (interpolación {lineal en numpy.percentile)

Dado un arreglo ordenado $x_{(1)}$, $x_{(2)}$,...,$x_{(n)}$  de tamaño $n$ y percentil $p$ en escala $[0,100]$

1. Se calcula la posición fraccional:

$$
h = (n - 1)\cdot \frac{p}{100}
$$

2. Luego:

- El índice inferior es:$i = \lfloor h \rfloor$

Nota: Piso o parte entera: ⌊⌋
Se llama función piso (o floor en inglés).
Significa “redondear hacia abajo al entero más cercano”. ⌊3.7⌋=3,

- La parte fraccional es:
$\delta=h-i$


````python
3. Finalmente, el percentil se interpola como:
````

$$
P_p = x_{(i)} + \delta \cdot \left( x_{(i+1)} - x_{(i)} \right)
$$

- Datos ordenados:
[40,55,60,70,72,75,78,80,85,90,95]

- Tamaño:n=11.

- Paso 1. Calcular posición fraccional:

h=(n−1)x (p/100)=(11−1)⋅0.75=10⋅0.75=7.5

-  Paso 2: Índices:

- i=⌊7.5⌋=7

- $\delta=h-i=7.5-7=0.5$

El elemento en posición 7 (recordando que Python indexa desde 0) es 80 y el siguiente (posición 8) es 85.

$percentil(75)=(1−0.5)⋅80+0.5⋅85=0.5⋅80+0.5⋅85=82.5$

Así, numpy.percentile devuelve 82.5 porque interpola entre el 8º y 9º valor.

Si ordenas los datos (ya lo están), el percentil 75 es la nota que marca el límite superior del 75% de los estudiantes con notas más bajas, y por encima quedan solo los mejores 25%.

En este caso, el percentil 75 ≈ 82.5. Eso significa:

-75% de los estudiantes sacaron 82.5 o menos.

-25% de los estudiantes sacaron más de 82.5


##### Resumen:
- Método “clásico” → percentil 75 = 85 (posición 9).
- 
- Método de numpy (con interpolación lineal) → percentil 75 = 82.5.

PERO — de todos estos componentes principales, ¿cuáles son lo suficientemente importantes como para conservar?

````python
import numpy as np
# Create a scree plot
plt.figure(figsize=(8,5))
plt.plot(range(1, len(pca.explained_variance_ratio_) + 1), pca.explained_variance_ratio_, marker='o', linestyle='--')
plt.title('Scree Plot')
plt.xlabel('Number of components')
plt.ylabel('Explained variance ratio')
plt.show()

# Calculate cumulative explained variance
cumulative_explained_variance = np.cumsum(pca.explained_variance_ratio_)

# Create a plot for cumulative explained variance
plt.figure(figsize=(8,5))
plt.plot(range(1, len(cumulative_explained_variance) + 1), cumulative_explained_variance, marker='o', linestyle='--')
plt.title('Cumulative Explained Variance')
plt.xlabel('Number of components')
plt.ylabel('Cumulative explained variance')
plt.axhline(y=0.9, color='r', linestyle='-')  # 90% variance line
plt.text(0.5, 0.85, '90% cut-off threshold', color = 'red', fontsize=16)
plt.show()
````

Basándonos en el gráfico de varianza explicada acumulada, normalmente se busca el número de componentes necesarios para superar el umbral elegido de varianza explicada — en este caso, el 90 %.

Según el gráfico proporcionado, parece que este umbral se supera entre el cuarto y el quinto componente. Dado que el gráfico muestra que con 4 componentes estamos justo por debajo del 90 % y que solo al agregar el quinto componente lo superamos, podemos decir:

- Al menos 5 componentes principales son importantes para capturar el 90 % de la varianza en los datos.
- Estos serían PC1, PC2, PC3, PC4 y PC5.

En la práctica, podrías optar por usar solo los primeros 4 componentes si estás dispuesto a aceptar ligeramente menos del 90 % de varianza explicada a cambio de un modelo más parsimonioso (es decir, un modelo con menos variables). Esta decisión puede depender del contexto específico y del equilibrio entre la simplicidad del modelo y la varianza explicada.

Opcional: Existe otra forma de determinar cuáles componentes principales son los más valiosos, conocida como el método del criterio de Kaiser.

````python
# Assuming pca is already fitted PCA object from sklearn
eigenvalues = pca.explained_variance_

# Applying Kaiser Criterion
kaiser_criterion = eigenvalues > 1

# Get the number of components that satisfy the Kaiser Criterion
n_components_kaiser = sum(kaiser_criterion)

# Print the components that satisfy the criterion
print(f"Number of components with eigenvalue > 1: {n_components_kaiser}")
important_components = np.arange(1, n_components_kaiser + 1)
print(f"Important components according to Kaiser Criterion: {important_components}")

# Optionally, if you want to see which specific components satisfy the criterion
for i, (ev, satisfy) in enumerate(zip(eigenvalues, kaiser_criterion), start=1):
    if satisfy:
        print(f"Component {i} with eigenvalue: {ev}")
````

##### Buenas Prácticas y Errores Comunes

Qué Hacer y Qué Evitar al Aplicar PCA

Haz:

- Estandariza tus datos antes de aplicar PCA, especialmente cuando las variables están medidas en diferentes escalas.
- Ten en cuenta el contexto del dominio al decidir cuántos componentes conservar; a veces, los componentes con menor varianza aún pueden tener un significado relevante.
- Utiliza un gráfico de sedimentación (scree plot) y la proporción de varianza acumulada para guiar la elección del número de componentes a mantener.


No Hacer:

- Ignorar la importancia de la selección de variables antes de aplicar PCA; no todas las variables pueden ser relevantes para el análisis.
- Pasar por alto la necesidad del conocimiento del dominio para interpretar correctamente los componentes principales.
- Confiar únicamente en PCA para la reducción de variables (Analisis Factorial)

##### Conceptos Erróneos Comunes

Concepto Erróneo: Más componentes significan un mejor modelo.
Realidad: A menudo, un número reducido de componentes puede proporcionar los conocimientos necesarios, y los componentes adicionales pueden solo añadir ruido.

Concepto Erróneo: El PCA puede manejar datos faltantes.
Realidad: El PCA requiere un conjunto de datos completo, por lo que debes tratar los valores faltantes antes de aplicarlo.

##### Consejos para un Uso Eficaz del PCA

- Correlación vs. Covarianza: Utiliza la matriz de covarianza cuando los datos estén en escalas similares, y la matriz de correlación cuando estén en escalas diferentes.
- Visualización: Usa gráficos de cargas (loadings) y de puntuaciones (scores) del PCA para visualizar y comprender la estructura de los datos.
- Enfoque Iterativo: Aplica PCA de forma iterativa y ajusta el número de componentes en función tanto de criterios estadísticos como de la calidad interpretativa.


##### Más Allá del PCA — Otras Técnicas de Reducción de Dimensionalidad

Aunque el PCA es una técnica poderosa, no es la única herramienta para la reducción de dimensionalidad.

- t-SNE (t-distributed Stochastic Neighbor Embedding): Es una técnica no lineal excelente para visualizar datos de alta dimensión en 2D o 3D. Usa t-SNE cuando las relaciones entre los puntos de datos son más complejas y el PCA no logra revelar estos patrones no lineales.
  
- LDA (Análisis Discriminante Lineal): A diferencia del PCA, el LDA es un método supervisado que se enfoca en maximizar la separabilidad entre distintas clases. Es más útil cuando se conocen las categorías del resultado.

### test de esfericidad de Barlett

El test de esfericidad de Bartlett es básicamente un portero para tu PCA o análisis factorial: ese decir, nos  dice si es apropiado seguir adelante o si tus datos no están listos para ese tipo de análisis.

rueba la hipótesis:

H₀ (nula): La matriz de correlaciones es la matriz identidad.Las variables no están correlacionadas → PCA/análisis factorial no tiene sentido.
H₁ (alternativa): La matriz de correlaciones es distinta de la identidad. Las variables sí están correlacionadas → Sí vale la pena aplicar PCA/análisis factorial.

Interpretación del resultado

- p < 0.05 → Rechazas H₀ → Hay correlaciones significativas → Puedes aplicar PCA.
- p ≥ 0.05 → No rechazas H₀ → Las variables no están suficientemente correlacionadas → PCA no aportará mucho.

````python
!pip install factor_analyzer

import pandas as pd
from factor_analyzer.factor_analyzer import calculate_bartlett_sphericity


chi_square_value, p_value = calculate_bartlett_sphericity(df)

print(f"Chi-cuadrado: {chi_square_value}")
print(f"p-valor: {p_value}")
````
