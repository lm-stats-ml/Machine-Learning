### Modelos CART (Classification and Regression Tree)

```
Los algoritmos basados en árboles se consideran uno de los mejores y más utilizados métodos de machine learning supervisado. Estos algoritmos dotan a los modelos predictivos de alta precisión, estabilidad y facilidad de interpretación. A diferencia de los modelos lineales, capturan bastante bien las relaciones no lineales. Son adaptables para resolver cualquier tipo de problema (clasificación o regresión).

Métodos como los árboles de decisión, random forest y gradient boosting se utilizan ampliamente en todo tipo de problemas de ciencia de datos. Por lo tanto, para todo analista (incluso principiantes), es importante aprender estos algoritmos y usarlos en la modelación.

Este notebook tiene como objetivo ayudar a los estudiantes a aprender los algoritmos basados en árboles desde cero. Al completar exitosamente este notebook, se espera que ustedes sean competentes en el uso de estos algoritmos y en la construcción de modelos predictivos.

Los árboles de decisión (DT) son un método de aprendizaje supervisado no paramétrico utilizado para la clasificación y la regresión.
El nodo superior de un árbol de decisión se conoce como nodo raíz, luego en cada bifurcación se divide según una variable explicativa o prdictora  y cada nodo final contiene una predicción para la variable objetivo. El objetivo es crear un modelo que prediga el valor de una variable objetivo mediante el aprendizaje de reglas de decisión simples inferidas a partir de las características de los datos. Un árbol puede considerarse una aproximación constante por partes.

No necesita StandardScaler. Los arboles no calculan distancias entre puntos, solo hacen preguntas binarias del tipo trip_distance > 3.5?. La escala de las variables es irrelevante para ese proceso. Parte 5 del codigo es exclusiva de CART. El toDebugString te imprime las reglas exactas en texto legible — algo que KNN y Logit no pueden hacer de esta forma. La importancia de variables ademas te dira cuanto aporta cada una de las tres al modelo, que es directamente comparable con los coeficientes beta del Logit.


Estructura de árbol: CART construye una estructura similar a un árbol compuesta por nodos y ramas.  En el árbol de decisión, los nodos se dividen en subnodos según un valor umbral de algún atributo. Cada bifurcación se divide según una variable predictora y cada nodo final contiene una predicción para la variable objetivo

El nodo raíz se toma del conjunto de entrenamiento y se divide en dos considerando el mejor atributo y su valor umbral. Luego, los subconjuntos también se dividen usando la misma lógica. Este proceso continúa hasta que se encuentra el último subconjunto puro en el árbol o hasta alcanzar el número máximo de hojas permitido durante el crecimiento del árbol.

El término CART se usa como un término genérico para las siguientes categorías de árboles de decisión:

- Árboles de Clasificación: El árbol se utiliza para determinar a qué “clase” es más probable que pertenezca la variable objetivo cuando esta es categórica.

- Árboles de Regresión: Se utilizan para predecir el valor de una variable continua.

#### Gini e Entropia. Una interpretacion intuitiva.
```
La analogia de la caja de canicas
Imagina que tienes una caja con canicas. Algunas son azules (pago con tarjeta) y otras son rojas (pago en efectivo).
La pregunta clave es: que tan "mezclada" esta la caja?

- Caja 1 — completamente mezclada:

50 canicas azules, 50 canicas rojas
Si metes la mano a ciegas, no tienes idea de que color vas a sacar.
Esta caja es completamente impura — Gini = 0.5, Entropia = 1.0

- Caja 2 — completamente pura:

100 canicas azules, 0 rojas
Si metes la mano, siempre sacaras azul — certeza total
Esta caja es completamente pura — Gini = 0.0, Entropia = 0.0

- Caja 3 — mayormente azul:

80 canicas azules, 20 rojas
Si metes la mano, probablemente sacaras azul — cierta certeza
Esta caja es bastante pura — Gini = 0.32, Entropia = 0.72

Gini = 1 - (0.80² + 0.20²) = 1 - (0.64 + 0.04) = 0.32

Entropia usa logaritmos (viene de la teoria de la informacion):

Entropia = -(p_tarjeta * log₂(p_tarjeta) + p_efectivo * log₂(p_efectivo))
         = -(0.80 * log₂(0.80) + 0.20 * log₂(0.20))
         = -(0.80 * (-0.322) + 0.20 * (-2.322))
         = -(-0.258 - 0.464) = 0.72


Tanto Gini como Entropia simplemente miden que tan mezclada esta la caja. Cuanto mas mezclada, mayor el numero. Cuanto mas pura, mas cerca de cero.

El objetivo de CART: separar las canicas

Cuando CART construye el arbol, su objetivo es hacer preguntas que separen las canicas mezcladas en cajas mas puras.
Volviendo al ejemplo de NYC Taxi — al inicio tienes todos los viajes juntos:

Caja inicial: 60% pagan con tarjeta, 40% en efectivo
Gini = 1 - (0.60² + 0.40²) = 1 - (0.36 + 0.16) = 0.48

Esta caja esta bastante mezclada. CART prueba todas las preguntas posibles y elige la que produzca las dos cajas resultantes mas puras.


Pregunta |1: fare_amount > 12? si la respuesta es SI vamos a la pregunta 2
Pregunta 2:  distancia > 3.5   si la respuesta es SI vamos a la pregunta 3
Pregunta 3: hora entre 7 y 22. Si la respuesta es si la p=0.80

Gini = 1 - (p_tarjeta² + p_efectivo²)
     = 1 - (0.80² + 0.20²)
     = 1 - (0.64 + 0.04) = 0.32

Entropia usa logaritmos (viene de la teoria de la informacion):

Entropia = -(p_tarjeta * log₂(p_tarjeta) + p_efectivo * log₂(p_efectivo))
         = -(0.80 * log₂(0.80) + 0.20 * log₂(0.20))
         = -(0.80 * (-0.322) + 0.20 * (-2.322))
         = -(-0.258 - 0.464) = 0.72

Ambas llegan a la misma conclusion: la caja con 80/20 es bastante pura. En la practica casi siempre eligen el mismo split. La diferencia es que Gini es mas rapida de calcular (no usa logaritmos) por eso PySpark la usa por defecto.


Pero esas probabilidades son de las hojas, no del nodo raiz. El p=0.80 de la hoja Tarjeta significa que el 80% de los viajes que llegan a esa hoja especifica (fare > 12, distancia > 3.5, hora entre 7 y 22) pagaron con tarjeta. No dice nada sobre cuantos viajes cumplen fare > 12 en total.


Resumen en una linea

Gini y Entropia son dos formas distintas de responder la misma pregunta: que tan mezclada esta la caja? CART las usa para encontrar la pregunta que mejor separa las canicas azules de las rojas. En PySpark usa Gini por defecto porque es mas eficiente — pero los resultados con Entropia son casi identicos.

````python
import pyspark.sql.functions as F

df = NYtaxidata2024 \
    .filter(F.col("payment_type").isin([1, 2])) \
    .filter(F.col("fare_amount") > 0) \
    .withColumn("label",
        F.when(F.col("payment_type") == 1, 1).otherwise(0))

# Distribucion global (la caja padre)
print("=== Distribucion global ===")
total = df.count()
df.groupBy("label").count() \
  .withColumn("proporcion", F.col("count") / total) \
  .show()

# Distribucion en cada rama del split fare_amount > 12
print("=== Rama SI: fare_amount > 12 ===")
rama_si = df.filter(F.col("fare_amount") > 12)
n_si = rama_si.count()
rama_si.groupBy("label").count() \
       .withColumn("proporcion", F.col("count") / n_si) \
       .show()

print("=== Rama NO: fare_amount <= 12 ===")
rama_no = df.filter(F.col("fare_amount") <= 12)
n_no = rama_no.count()
rama_no.groupBy("label").count() \
       .withColumn("proporcion", F.col("count") / n_no) \
       .show()
````

````python
La gran mayoria paga con tarjeta en NYC Taxi 2024
Dado que tengo  p_tarjeta y p_efectivo para cada rama, calculo  Gini directamente:
````

````python
# Formula Gini manual con los valores reales
def gini(p_tarjeta):
    p_efectivo = 1 - p_tarjeta
    return 1 - (p_tarjeta**2 + p_efectivo**2)

# Ejemplo con valores hipoteticos — reemplaza con los tuyos
p_tarjeta_padre  = 0.8434   # reemplazar con resultado real
p_tarjeta_si     = 0.8508   # reemplazar con resultado real
p_tarjeta_no     = 0.8342   # reemplazar con resultado real
peso_si          = n_si / total
peso_no          = n_no / total

gini_padre = gini(p_tarjeta_padre)
gini_si    = gini(p_tarjeta_si)
gini_no    = gini(p_tarjeta_no)
ganancia   = gini_padre - (gini_si * peso_si + gini_no * peso_no)

print(f"Gini padre : {gini_padre:.4f}")
print(f"Gini SI    : {gini_si:.4f}")
print(f"Gini NO    : {gini_no:.4f}")
print(f"Ganancia   : {ganancia:.4f}")
````

**Gini rama SI (fare > 12):**

**Gini rama NO (fare <= 12):**

La conclusion importante
El split fare_amount > 12 casi no aporta ganancia de informacion con estos datos reales. El motivo es claro al ver los resultados:

Rama SI: 85.1% tarjeta
Rama NO: 83.4% tarjeta

Las dos ramas son casi identicas. Dividir por fare_amount > 12 no separa bien los pagos con tarjeta de los pagos en efectivo — ambas ramas quedan igual de "mezcladas".

Segundo, y mas importante: esto sugiere que fare_amount > 12 no es el mejor primer split para predecir metodo de pago. El arbol real de PySpark probablemente elegiria otra variable para el nodo raiz — posiblemente trip_distance, PULocationID o hour — porque esas variables generan mayor separacion entre tarjeta y efectivo que el monto de la tarifa.

### Pronostícando la probabilidad de pagar con tarjeta

````python
# ══════════════════════════════════════════════════════════════
# CART — PREDICCIÓN MÉTODO DE PAGO (Tarjeta=1 vs Efectivo=0)
# Variables: trip_distance, tolls_amount, passenger_count
# Comparacion directa con KNN y Logit previos
# ══════════════════════════════════════════════════════════════

import pyspark.sql.functions as F
from pyspark.ml import Pipeline
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.classification import DecisionTreeClassifier
from pyspark.ml.evaluation import (
    BinaryClassificationEvaluator,
    MulticlassClassificationEvaluator
)
from pyspark.ml.tuning import ParamGridBuilder, CrossValidator

# Las MISMAS 3 variables del Logit y KNN
feature_cols = [
    "trip_distance",    # β1 = -0.0607 *
    "tolls_amount",     # β3 = +0.0836 ***
    "passenger_count",  # β4 = -0.0880 ***
]

# ════════════════════════════════════════════════════════════════
# PARTE 1 — PREPARACIÓN DE DATOS
# Mismos filtros que Logit y KNN para comparación justa
# ════════════════════════════════════════════════════════════════

df_cart = NYtaxidata2024 \
    .filter(F.col("payment_type").isin([1, 2])) \
    .filter(F.col("trip_distance")   >  0) \
    .filter(F.col("tolls_amount")    >= 0) \
    .filter(F.col("passenger_count") >  0) \
    .withColumn("label",
        F.when(F.col("payment_type") == 1, 1.0)
         .otherwise(0.0)) \
    .dropna(subset=feature_cols + ["label"])

print("=== Distribucion de la variable objetivo ===")
total = df_cart.count()
df_cart.groupBy("label") \
       .count() \
       .withColumn("Metodo",
           F.when(F.col("label") == 1.0, "Tarjeta (1)")
            .otherwise("Efectivo (0)")) \
       .withColumn("Proporcion",
           F.round(F.col("count") / total, 4)) \
       .select("Metodo", "count", "Proporcion") \
       .show()

# ════════════════════════════════════════════════════════════════
# PARTE 2 — PIPELINE PYSPARK
# CART NO requiere StandardScaler — ventaja sobre KNN y SVM
# ════════════════════════════════════════════════════════════════

assembler = VectorAssembler(
    inputCols=feature_cols,
    outputCol="features",
    handleInvalid="skip"
)

dt = DecisionTreeClassifier(
    featuresCol="features",
    labelCol="label",
    maxDepth=5,           # punto de partida — se optimiza con CV
    impurity="gini",      # criterio de division
    minInstancesPerNode=20,
    seed=42
)

pipeline = Pipeline(stages=[assembler, dt])

# ════════════════════════════════════════════════════════════════
# PARTE 3 — ENTRENAMIENTO
# Mismo random split que KNN para comparacion justa
# ════════════════════════════════════════════════════════════════

df_train, df_test = df_cart.randomSplit([0.8, 0.2], seed=42)

print(f"Train: {df_train.count():,} | Test: {df_test.count():,}")

model = pipeline.fit(df_train)
preds = model.transform(df_test)

# ════════════════════════════════════════════════════════════════
# PARTE 4 — EVALUACION
# ════════════════════════════════════════════════════════════════

auc = BinaryClassificationEvaluator(
    labelCol="label",
    metricName="areaUnderROC"
).evaluate(preds)

# Nombres correctos en PySpark MLlib
metricas = {
    "accuracy"  : "accuracy",
    "f1"        : "f1",
    "precision" : "weightedPrecision",   # <-- no "precision"
    "recall"    : "weightedRecall"       # <-- no "recall"
}

resultados = {}
for nombre, metrica_spark in metricas.items():
    val = MulticlassClassificationEvaluator(
        labelCol="label",
        predictionCol="prediction",
        metricName=metrica_spark
    ).evaluate(preds)
    resultados[nombre] = val

print("\n=== RESULTADOS CART (maxDepth=5) ===")
print(f"AUC-ROC   : {auc:.4f}")
for nombre, val in resultados.items():
    print(f"{nombre:10s}: {val:.4f}")

# ════════════════════════════════════════════════════════════════
# PARTE 5 — INTERPRETACION DEL ARBOL
# Ventaja unica de CART frente a KNN y Logit
# ════════════════════════════════════════════════════════════════

dt_model = model.stages[-1]

print("\n=== ESTRUCTURA DEL ARBOL ===")
print(f"Profundidad real : {dt_model.depth}")
print(f"Numero de nodos  : {dt_model.numNodes}")

print("\n=== IMPORTANCIA DE VARIABLES ===")
print(f"{'Variable':20s}  {'Importancia':>12}  {'Interpretacion'}")
print("-" * 60)
for name, imp in zip(feature_cols, dt_model.featureImportances):
    barra = "#" * int(imp * 40)
    print(f"{name:20s}  {imp:>12.4f}  {barra}")

print("\n=== REGLAS DEL ARBOL (toDebugString) ===")
print(dt_model.toDebugString)

# ════════════════════════════════════════════════════════════════
# PARTE 6 — MATRIZ DE CONFUSION
# ════════════════════════════════════════════════════════════════

print("\n=== MATRIZ DE CONFUSION ===")
preds.groupBy("label", "prediction") \
     .count() \
     .orderBy("label", "prediction") \
     .withColumn("Clase_Real",
         F.when(F.col("label") == 1.0, "Tarjeta")
          .otherwise("Efectivo")) \
     .withColumn("Clase_Pred",
         F.when(F.col("prediction") == 1.0, "Tarjeta")
          .otherwise("Efectivo")) \
     .select("Clase_Real", "Clase_Pred", "count") \
     .show()

# ════════════════════════════════════════════════════════════════
# PARTE 7 — OPTIMIZACION DE maxDepth CON CROSS VALIDATION
# ════════════════════════════════════════════════════════════════

print("\n=== CROSS VALIDATION — Buscando maxDepth optimo ===")

grid = ParamGridBuilder() \
    .addGrid(dt.maxDepth,  [3, 5, 7]) \
    .addGrid(dt.impurity,  ["gini", "entropy"]) \
    .build()

# 3 folds x 6 combinaciones = 18 entrenamientos
cv = CrossValidator(
    estimator=pipeline,
    estimatorParamMaps=grid,
    evaluator=BinaryClassificationEvaluator(
        labelCol="label",
        metricName="areaUnderROC"
    ),
    numFolds=3,
    seed=42
)

cv_model = cv.fit(df_train)
best_model = cv_model.bestModel
best_dt    = best_model.stages[-1]

print(f"\nMejor maxDepth : {best_dt.getMaxDepth()}")
print(f"Mejor impurity : {best_dt.getImpurity()}")

# Evaluar el mejor modelo en test
best_preds = best_model.transform(df_test)
best_auc   = BinaryClassificationEvaluator(
    labelCol="label",
    metricName="areaUnderROC"
).evaluate(best_preds)

print(f"AUC-ROC (mejor modelo): {best_auc:.4f}")

# ════════════════════════════════════════════════════════════════
# PARTE 8 — TABLA COMPARATIVA CON LOGIT Y KNN
# ════════════════════════════════════════════════════════════════

print("\n" + "="*55)
print("COMPARACION FINAL — mismas variables, mismo dataset")
print("="*55)
print(f"{'Modelo':15s}  {'AUC-ROC':>8}  {'Accuracy':>9}  {'Notas'}")
print("-"*55)
print(f"{'Logit':15s}  {'0.5196':>8}  {'0.6619':>9}  coeficientes interpretables")
print(f"{'KNN (K=5)':15s}  {'0.5344':>8}  {'0.8420':>9}  muestra 5%, no escala")
print(f"{'CART (CV)':15s}  {best_auc:>8.4f}  {'0.8435':>9}  reglas legibles, full data")
print("="*55)
print("Reemplaza los ? con los valores de tus modelos anteriores")
````
