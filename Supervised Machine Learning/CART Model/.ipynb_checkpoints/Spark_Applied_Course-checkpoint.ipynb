````python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, lit

# --------------------------------------------------
# 1. Spark session (si ya existe, puedes omitir esto)
# --------------------------------------------------
spark = SparkSession.builder.getOrCreate()

# --------------------------------------------------
# 2. Describe del DataFrame
# --------------------------------------------------
desc = df1.describe()

# --------------------------------------------------
# 3. Transponer describe(): estadísticas por variable
#    Resultado: variable | statistic | value
# --------------------------------------------------
variables = desc.columns[1:]  # quitamos la columna "summary"

desc_long = None

for v in variables:
    tmp = (
        desc
        .select(
            lit(v).alias("variable"),
            col("summary").alias("statistic"),
            col(v).alias("value")
        )
    )
    
    desc_long = tmp if desc_long is None else desc_long.union(tmp)

# --------------------------------------------------
# 4. (Opcional) ordenar para que quede bonito
# --------------------------------------------------
desc_long = desc_long.orderBy("variable", "statistic")

# --------------------------------------------------
# 5. Exportar a CSV en un solo archivo
# --------------------------------------------------
(
    desc_long
    .coalesce(1)
    .write
    .option("header", True)
    .mode("overwrite")
    .csv("./outputs/describe_df1")
)

print("✔ Describe transpuesto y exportado correctamente")
````

#### DataFrames y SQL

Ahora podemos consultar nuestros datos en SQL. Para hacerlo, usaremos la función spark.sql (recuerda, spark es nuestra variable SparkSession) que convenientemente devuelve un nuevo DataFrame. Aunque esto podría parecer un poco circular en lógica—que una consulta SQL contra un DataFrame devuelve otro DataFrame—en realidad es bastante poderoso

````python
# Crear la vista temporal con el nombre correcto del DataFrame
NYtaxidata2024.createOrReplaceTempView("nyc_taxi_data")

# Forma SQL
sqlWay = spark.sql("""
SELECT PULocationID, count(1) as trip_count
FROM nyc_taxi_data
GROUP BY PULocationID
""")

# Forma DataFrame API - usando el nombre correcto
dataFrameWay = NYtaxidata2024 \
    .groupBy("PULocationID") \
    .count()

# Mostrar los planes de ejecución
sqlWay.explain()
dataFrameWay.explain()
````

````python
# 1. Por zona de destino (Drop-off)
sqlWay = spark.sql("""
SELECT DOLocationID, count(1) as trip_count
FROM nyc_taxi_data
GROUP BY DOLocationID
""")

dataFrameWay = NYtaxidata2024.groupBy("DOLocationID").count()

# 2. Por tipo de pago
sqlWay = spark.sql("""
SELECT payment_type, count(1) as trip_count
FROM nyc_taxi_data
GROUP BY payment_type
""")

dataFrameWay = NYtaxidata2024.groupBy("payment_type").count()

# 3. Por número de pasajeros
sqlWay = spark.sql("""
SELECT passenger_count, count(1) as trip_count
FROM nyc_taxi_data
WHERE passenger_count IS NOT NULL
GROUP BY passenger_count
ORDER BY passenger_count
""")

dataFrameWay = NYtaxidata2024 \
    .filter("passenger_count IS NOT NULL") \
    .groupBy("passenger_count") \
    .count() \
    .orderBy("passenger_count")
````

````python
dataFrameWay.explain()
````

````python
# Tu DataFrame ya cargado
NYtaxidata2024 = spark.read \
    .parquet("../data/Database/nyc_taxi_data/yellow_tripdata_2024-01.parquet")

# Crear la vista temporal
NYtaxidata2024.createOrReplaceTempView("nyc_taxi_data")

# Forma SQL
sqlWay = spark.sql("""
SELECT PULocationID, count(1)
FROM nyc_taxi_data
GROUP BY PULocationID
""")

# Forma DataFrame API
dataFrameWay = NYtaxidata2024 \
    .groupBy("PULocationID") \
    .count()

# Mostrar los planes de ejecución para comprobar que son equivalentes
print("=" * 80)
print("PLAN DE EJECUCIÓN - SQL WAY")
print("=" * 80)
sqlWay.explain()

print("\n" + "=" * 80)
print("PLAN DE EJECUCIÓN - DATAFRAME API WAY")
print("=" * 80)
dataFrameWay.explain()

# Opcional: Verificar que ambos producen los mismos resultados
print("\n" + "=" * 80)
print("COMPARACIÓN DE RESULTADOS")
print("=" * 80)
print("SQL Way - Primeras 5 filas:")
sqlWay.show(5)

print("\nDataFrame API Way - Primeras 5 filas:")
dataFrameWay.show(5)

# Opcional: Verificar que el conteo total es el mismo
print(f"\nTotal de filas SQL Way: {sqlWay.count()}")
print(f"Total de filas DataFrame Way: {dataFrameWay.count()}")


# **Lo que deberías observar:**

#1. **Planes de ejecución idénticos**: Ambos métodos producirán el mismo plan físico
#2. **Misma optimización**: Catalyst optimizer produce el mismo plan optimizado
#3. **Mismo rendimiento**: No hay diferencia de performance entre SQL y DataFrame API

#**Ejemplo esperado del plan de ejecución:**
#```
#== Physical Plan ==
#AdaptiveSparkPlan
#+- HashAggregate(keys=[PULocationID#...], functions=[count(1)])
#   +- Exchange hashpartitioning(PULocationID#..., 200)
#      +- HashAggregate(keys=[PULocationID#...], functions=[partial_count(1)])
#         +- FileScan parquet [PULocationID#...] ...
````

#### 4. Análisis de Eficiencia y Calidad 

````python
# A. Análisis de propinas por método de pago
spark.sql("""
SELECT payment_type,
       count(1) as trips,
       round(avg(tip_amount), 2) as avg_tip,
       round(avg(tip_amount/total_amount * 100), 2) as tip_percentage
FROM nyc_taxi_data 
WHERE total_amount > 0
GROUP BY payment_type
ORDER BY payment_type
""").show()

# B. Eficiencia: tarifa por milla
spark.sql("""
SELECT 
    CASE 
        WHEN trip_distance <= 1 THEN '0-1 millas'
        WHEN trip_distance <= 5 THEN '1-5 millas'
        WHEN trip_distance <= 10 THEN '5-10 millas'
        ELSE '10+ millas'
    END as distance_range,
    count(1) as trips,
    round(avg(total_amount/trip_distance), 2) as avg_per_mile,
    round(avg(trip_distance), 2) as avg_distance
FROM nyc_taxi_data 
WHERE trip_distance > 0
GROUP BY 
    CASE 
        WHEN trip_distance <= 1 THEN '0-1 millas'
        WHEN trip_distance <= 5 THEN '1-5 millas'
        WHEN trip_distance <= 10 THEN '5-10 millas'
        ELSE '10+ millas'
    END
""").show()
````

````python
# PROBLEMA: timestamp_ntz no es útil directamente para ML
# SOLUCIÓN: Extraer características temporales

from pyspark.sql.functions import hour, dayofweek, month, year, weekofyear

df_ml = NYtaxidata2024 \
    .withColumn("pickup_hour", hour("tpep_pickup_datetime")) \
    .withColumn("pickup_dayofweek", dayofweek("tpep_pickup_datetime")) \
    .withColumn("pickup_month", month("tpep_pickup_datetime")) \
    .withColumn("pickup_weekofyear", weekofyear("tpep_pickup_datetime")) \
    .withColumn("is_weekend", (dayofweek("tpep_pickup_datetime").isin([1, 7])).cast("int"))

# Calcular duración del viaje
from pyspark.sql.functions import unix_timestamp

df_ml = df_ml.withColumn(
    "trip_duration_minutes",
    (unix_timestamp("tpep_dropoff_datetime") - unix_timestamp("tpep_pickup_datetime")) / 60
)
````

##### 3. Valores Nulos (nullable = true en TODAS las columnas)

````python
from pyspark.sql.functions import col, sum, count, lit, round as spark_round

# Obtener total de filas
total_rows = NYtaxidata2024.count()

# Crear resumen de nulos
null_analysis = NYtaxidata2024.select([
    sum(col(c).isNull().cast("int")).alias(f"{c}_count")
    for c in NYtaxidata2024.columns
])

# Transponer para mejor visualización
null_data = null_analysis.collect()[0].asDict()

# Crear DataFrame de resumen
from pyspark.sql.types import StructType, StructField, StringType, LongType, DoubleType

schema = StructType([
    StructField("column_name", StringType(), True),
    StructField("null_count", LongType(), True),
    StructField("null_percentage", DoubleType(), True)
])

null_summary_data = [
    (col_name, null_count, round((null_count / total_rows) * 100, 2))
    for col_name, null_count in null_data.items()
]

null_summary_df = spark.createDataFrame(null_summary_data, schema)
null_summary_df.orderBy("null_percentage", ascending=False).show(20, truncate=False)
````

````python
from pyspark.sql.functions import col, abs as spark_abs

print("="*80)
print("LIMPIEZA AGRESIVA - ELIMINAR OUTLIERS EXTREMOS")
print("="*80)

# PASO 1: Estadísticas para identificar límites razonables
print("\n Calculando percentiles para límites razonables...")

# Calcular percentiles en datos SIN nulos
percentiles = NYtaxidata2024.filter(
    col("RatecodeID").isNotNull()
).approxQuantile(
    ["trip_distance", "total_amount"],
    [0.01, 0.99],  # 1% y 99%
    0.01
)

distance_p01, distance_p99 = percentiles[0]
amount_p01, amount_p99 = percentiles[1]

print(f"trip_distance - P01: {distance_p01:.2f}, P99: {distance_p99:.2f}")
print(f"total_amount - P01: {amount_p01:.2f}, P99: {amount_p99:.2f}")

# PASO 2: Aplicar limpieza AGRESIVA
df_clean_agresivo = NYtaxidata2024.filter(
    # A. Eliminar nulos en campos críticos
    col("RatecodeID").isNotNull() &
    col("store_and_fwd_flag").isNotNull() &
    col("congestion_surcharge").isNotNull() &
    col("passenger_count").isNotNull() &
    col("PULocationID").isNotNull() &
    col("DOLocationID").isNotNull() &
    col("tpep_pickup_datetime").isNotNull() &
    col("tpep_dropoff_datetime").isNotNull() &
    
    # B. Filtrar valores razonables (basado en conocimiento del dominio)
    (col("trip_distance") > 0) & (col("trip_distance") < 50) &  # Max 50 millas (razonable para NYC)
    (col("total_amount") > 0) & (col("total_amount") < 300) &   # Max $300 (P99 ≈ $70-80)
    (col("fare_amount") > 0) & (col("fare_amount") < 200) &
    (col("passenger_count").between(1, 6)) &
    (col("VendorID").isin([1, 2])) &
    
    # C. Eliminar tarifas negativas (reembolsos/errores)
    (col("tip_amount") >= 0) &
    (col("tolls_amount") >= 0) &
    (col("extra") >= 0)
)

# PASO 3: Reportar resultados
total_original = NYtaxidata2024.count()
total_limpio = df_clean_agresivo.count()
eliminados = total_original - total_limpio

print(f"\n RESULTADOS DE LIMPIEZA AGRESIVA:")
print(f"   Registros originales:  {total_original:,}")
print(f"   Registros limpios:     {total_limpio:,}")
print(f"   Registros eliminados:  {eliminados:,} ({(eliminados/total_original)*100:.2f}%)")
print(f"   Tasa de retención:     {(total_limpio/total_original)*100:.2f}%")

print("\n" + "="*80)
````

#### Codigo final

````python
from pyspark.sql.functions import col, unix_timestamp, hour, dayofweek, month, when

print("="*100)
print("🚀 LIMPIEZA AGRESIVA COMPLETA - NYC TAXI DATA PARA MACHINE LEARNING")
print("="*100)

# =====================================================================
# PASO 1: CALCULAR PERCENTILES PARA LÍMITES RAZONABLES
# =====================================================================
print("\n📊 PASO 1: Calculando percentiles para establecer límites...")

# Calcular percentiles en datos sin nulos críticos
data_para_percentiles = NYtaxidata2024.filter(
    col("RatecodeID").isNotNull() &
    (col("trip_distance") > 0) &
    (col("total_amount") > 0)
)

percentiles = data_para_percentiles.approxQuantile(
    ["trip_distance", "total_amount"],
    [0.01, 0.25, 0.50, 0.75, 0.95, 0.99],
    0.01
)

print("\n📈 Percentiles de trip_distance:")
print(f"   P01: {percentiles[0][0]:.2f} millas")
print(f"   P25: {percentiles[0][1]:.2f} millas")
print(f"   P50 (Mediana): {percentiles[0][2]:.2f} millas")
print(f"   P75: {percentiles[0][3]:.2f} millas")
print(f"   P95: {percentiles[0][4]:.2f} millas")
print(f"   P99: {percentiles[0][5]:.2f} millas")

print("\n💰 Percentiles de total_amount:")
print(f"   P01: ${percentiles[1][0]:.2f}")
print(f"   P25: ${percentiles[1][1]:.2f}")
print(f"   P50 (Mediana): ${percentiles[1][2]:.2f}")
print(f"   P75: ${percentiles[1][3]:.2f}")
print(f"   P95: ${percentiles[1][4]:.2f}")
print(f"   P99: ${percentiles[1][5]:.2f}")

# =====================================================================
# PASO 2: APLICAR LIMPIEZA AGRESIVA
# =====================================================================
print("\n" + "="*100)
print("🧹 PASO 2: Aplicando filtros de limpieza...")

total_original = NYtaxidata2024.count()
print(f"\n📊 Registros originales: {total_original:,}")

# Aplicar filtros progresivamente y reportar
print("\n🔍 Aplicando filtros:")

# Filtro 1: Eliminar nulos críticos
df_step1 = NYtaxidata2024.filter(
    col("RatecodeID").isNotNull() &
    col("store_and_fwd_flag").isNotNull() &
    col("congestion_surcharge").isNotNull() &
    col("passenger_count").isNotNull() &
    col("PULocationID").isNotNull() &
    col("DOLocationID").isNotNull() &
    col("tpep_pickup_datetime").isNotNull() &
    col("tpep_dropoff_datetime").isNotNull()
)
count_step1 = df_step1.count()
print(f"   ✓ Después de eliminar nulos críticos:        {count_step1:,} ({((total_original-count_step1)/total_original)*100:.2f}% eliminado)")

# Filtro 2: Valores razonables de distancia
df_step2 = df_step1.filter(
    (col("trip_distance") > 0) & (col("trip_distance") < 50)  # Max 50 millas razonable para NYC
)
count_step2 = df_step2.count()
print(f"   ✓ Después de filtrar distancias (0-50 mi):    {count_step2:,} ({((count_step1-count_step2)/count_step1)*100:.2f}% eliminado)")

# Filtro 3: Valores razonables de tarifa
df_step3 = df_step2.filter(
    (col("total_amount") > 2.50) & (col("total_amount") < 300)  # Min tarifa base, max razonable
)
count_step3 = df_step3.count()
print(f"   ✓ Después de filtrar tarifas ($2.50-$300):    {count_step3:,} ({((count_step2-count_step3)/count_step2)*100:.2f}% eliminado)")

# Filtro 4: Pasajeros válidos
df_step4 = df_step3.filter(
    col("passenger_count").between(1, 6)
)
count_step4 = df_step4.count()
print(f"   ✓ Después de filtrar pasajeros (1-6):         {count_step4:,} ({((count_step3-count_step4)/count_step3)*100:.2f}% eliminado)")

# Filtro 5: Vendors válidos
df_step5 = df_step4.filter(
    col("VendorID").isin([1, 2])
)
count_step5 = df_step5.count()
print(f"   ✓ Después de filtrar vendors (1,2):           {count_step5:,} ({((count_step4-count_step5)/count_step4)*100:.2f}% eliminado)")

# Filtro 6: Sin valores negativos
df_clean = df_step5.filter(
    (col("fare_amount") > 0) &
    (col("tip_amount") >= 0) &
    (col("tolls_amount") >= 0) &
    (col("extra") >= 0)
)
count_clean = df_clean.count()
print(f"   ✓ Después de eliminar valores negativos:      {count_clean:,} ({((count_step5-count_clean)/count_step5)*100:.2f}% eliminado)")

# =====================================================================
# PASO 3: FEATURE ENGINEERING
# =====================================================================
print("\n" + "="*100)
print("🔧 PASO 3: Creando features para Machine Learning...")

df_ml = df_clean \
    .withColumn("pickup_hour", hour("tpep_pickup_datetime")) \
    .withColumn("pickup_dayofweek", dayofweek("tpep_pickup_datetime")) \
    .withColumn("pickup_month", month("tpep_pickup_datetime")) \
    .withColumn("is_weekend", when(dayofweek("tpep_pickup_datetime").isin([1, 7]), 1).otherwise(0)) \
    .withColumn("trip_duration_minutes",
                (unix_timestamp("tpep_dropoff_datetime") - 
                 unix_timestamp("tpep_pickup_datetime")) / 60) \
    .withColumn("speed_mph", 
                col("trip_distance") / (col("trip_duration_minutes") / 60))

# Filtrar duraciones razonables
df_ml = df_ml.filter(
    (col("trip_duration_minutes") > 1) &  # Mínimo 1 minuto
    (col("trip_duration_minutes") < 180)  # Máximo 3 horas
)

count_ml = df_ml.count()
print(f"   ✓ Después de filtrar duraciones (1-180 min):  {count_ml:,}")

# Imputar solo campos opcionales
df_ml = df_ml.na.fill({
    "Airport_fee": 0.0
})

print(f"\n✅ Features creadas:")
print(f"   - pickup_hour (0-23)")
print(f"   - pickup_dayofweek (1=Domingo, 7=Sábado)")
print(f"   - pickup_month")
print(f"   - is_weekend (0/1)")
print(f"   - trip_duration_minutes")
print(f"   - speed_mph")

# =====================================================================
# PASO 4: REPORTE DE RESULTADOS
# =====================================================================
print("\n" + "="*100)
print("📊 PASO 4: REPORTE FINAL DE LIMPIEZA")
print("="*100)

total_final = df_ml.count()
eliminados = total_original - total_final
retencion = (total_final / total_original) * 100

print(f"\n📈 RESUMEN ESTADÍSTICO:")
print(f"   Registros originales:      {total_original:>12,}")
print(f"   Registros finales:         {total_final:>12,}")
print(f"   Registros eliminados:      {eliminados:>12,}")
print(f"   Porcentaje eliminado:      {((eliminados/total_original)*100):>11.2f}%")
print(f"   Tasa de retención:         {retencion:>11.2f}%")

# =====================================================================
# PASO 5: VERIFICACIÓN DE CALIDAD
# =====================================================================
print("\n" + "="*100)
print("✅ PASO 5: VERIFICACIÓN DE CALIDAD DEL DATASET LIMPIO")
print("="*100)

print("\n🔍 Verificación de NULOS:")
for col_name in ["RatecodeID", "passenger_count", "PULocationID", "DOLocationID", 
                 "trip_distance", "total_amount", "pickup_hour", "trip_duration_minutes"]:
    nulos = df_ml.filter(col(col_name).isNull()).count()
    status = "✅" if nulos == 0 else "❌"
    print(f"   {status} {col_name:30} | Nulos: {nulos}")

print("\n🔍 Verificación de OUTLIERS:")
outliers_checks = [
    ("Viajes > 50 millas", df_ml.filter(col("trip_distance") > 50).count()),
    ("Tarifas > $300", df_ml.filter(col("total_amount") > 300).count()),
    ("Tarifas < $2.50", df_ml.filter(col("total_amount") < 2.50).count()),
    ("Pasajeros > 6", df_ml.filter(col("passenger_count") > 6).count()),
    ("Pasajeros < 1", df_ml.filter(col("passenger_count") < 1).count()),
    ("Distancia = 0", df_ml.filter(col("trip_distance") == 0).count()),
    ("Duración > 180 min", df_ml.filter(col("trip_duration_minutes") > 180).count()),
    ("Valores negativos", df_ml.filter(
        (col("total_amount") < 0) | 
        (col("tip_amount") < 0) | 
        (col("fare_amount") < 0)
    ).count())
]

for check_name, count_val in outliers_checks:
    status = "✅" if count_val == 0 else "⚠️"
    print(f"   {status} {check_name:30} | {count_val:>10,}")

# =====================================================================
# PASO 6: ESTADÍSTICAS DESCRIPTIVAS
# =====================================================================
print("\n" + "="*100)
print("📊 PASO 6: ESTADÍSTICAS DESCRIPTIVAS DEL DATASET LIMPIO")
print("="*100)

print("\n📈 Estadísticas de variables principales:")
df_ml.select(
    "trip_distance", "total_amount", "passenger_count", 
    "trip_duration_minutes", "speed_mph"
).describe().show()

print("\n📊 Distribución por VendorID:")
df_ml.groupBy("VendorID").count().orderBy("count", ascending=False).show()

print("\n📊 Distribución por RatecodeID:")
df_ml.groupBy("RatecodeID").count().orderBy("count", ascending=False).show()

print("\n📊 Distribución por payment_type:")
df_ml.groupBy("payment_type").count().orderBy("count", ascending=False).show()

print("\n📊 Top 10 zonas de recogida (PULocationID):")
df_ml.groupBy("PULocationID").count().orderBy("count", ascending=False).limit(10).show()

print("\n📊 Distribución por hora del día:")
df_ml.groupBy("pickup_hour").count().orderBy("pickup_hour").show(24)

print("\n📊 Distribución por día de la semana:")
df_ml.groupBy("pickup_dayofweek").count().orderBy("pickup_dayofweek").show()

# =====================================================================
# PASO 7: GUARDAR DATASET LIMPIO
# =====================================================================
print("\n" + "="*100)
print("💾 PASO 7: GUARDANDO DATASET LIMPIO")
print("="*100)

output_path = "../data/nyc_taxi_ml_ready.parquet"
print(f"\n💾 Guardando en: {output_path}")

df_ml.write.mode("overwrite").parquet(output_path)

print(f"✅ Dataset guardado exitosamente!")
print(f"   Registros: {df_ml.count():,}")
print(f"   Columnas: {len(df_ml.columns)}")

# =====================================================================
# PASO 8: RESUMEN FINAL
# =====================================================================
print("\n" + "="*100)
print("🎉 RESUMEN FINAL - DATASET LISTO PARA MACHINE LEARNING")
print("="*100)

print(f"""
✅ CALIDAD DEL DATASET:
   - Sin valores nulos en campos críticos
   - Sin outliers extremos
   - Sin valores negativos (excepto reembolsos válidos)
   - Valores dentro de rangos razonables para NYC
   - Features temporales creadas
   - Duración de viaje calculada
   - Velocidad promedio calculada

📊 CARACTERÍSTICAS:
   - Total de registros: {df_ml.count():,}
   - Tasa de retención: {retencion:.2f}%
   - Columnas totales: {len(df_ml.columns)}
   - Features nuevas: 6 (pickup_hour, pickup_dayofweek, pickup_month, 
                         is_weekend, trip_duration_minutes, speed_mph)

🎯 PRÓXIMOS PASOS:
   1. Encoding de variables categóricas (VendorID, RatecodeID, payment_type)
   2. Feature scaling (StandardScaler o MinMaxScaler)
   3. Train/Test split (80/20 o 70/30)
   4. Entrenamiento de modelo ML
   
💾 DATASET GUARDADO EN: {output_path}
""")

print("="*100)
print("✅ LIMPIEZA COMPLETA FINALIZADA")
print("="*100)
````

##### Balanceo muestra

````python
print("="*100)
print("🚀 IMPLEMENTACIÓN COMPLETA - ESTRATEGIA HÍBRIDA RECOMENDADA")
print("="*100)

from pyspark.sql.functions import col, when

# Cargar datos
train_data = spark.read.parquet("../data/train_data.parquet")
test_data = spark.read.parquet("../data/test_data.parquet")

# Filtrar solo payment_type 1 y 2
train = train_data.filter(col("payment_type").isin([1, 2]))
test = test_data.filter(col("payment_type").isin([1, 2]))

# Separar clases
class_1 = train.filter(col("payment_type") == 1)
class_2 = train.filter(col("payment_type") == 2)

# Undersampling a ratio 2.5:1
target = int(class_2.count() * 2.5)
fraction = target / class_1.count()
class_1_sampled = class_1.sample(False, fraction, seed=42)
train_balanced = class_1_sampled.union(class_2)

# Calcular class weights
total = train_balanced.count()
c1 = train_balanced.filter(col("payment_type") == 1).count()
c2 = train_balanced.filter(col("payment_type") == 2).count()

w1 = total / (2 * c1)
w2 = total / (2 * c2)

train_final = train_balanced.withColumn(
    "class_weight",
    when(col("payment_type") == 1, w1).otherwise(w2)
)

# Guardar
train_final.write.mode("overwrite").parquet("../data/train_balanced_final.parquet")
test.write.mode("overwrite").parquet("../data/test_balanced_final.parquet")

print(f"""
✅ COMPLETADO!

📊 Resultados:
   Train original:    {train.count():,}
   Train balanceado:  {train_final.count():,}
   Ratio final:       {c1/c2:.1f}:1
   
💾 Archivos guardados:
   - train_balanced_final.parquet
   - test_balanced_final.parquet
   
🎯 Próximo paso:
   Entrenar modelo con weightCol='class_weight'
""")
````

##### PIPELINE COMPLETO. Un Pipeline encadena múltiples transformaciones en un flujo secuencial, ejecutándolas en orden automáticamente.

##### 

````python
from pyspark.sql.functions import col, when
from pyspark.ml.feature import VectorAssembler, StringIndexer
from pyspark.ml.classification import LogisticRegression
from pyspark.ml.evaluation import BinaryClassificationEvaluator, MulticlassClassificationEvaluator

print("="*100)
print(" PIPELINE COMPLETO - ESTRATEGIA RECOMENDADA CON MÉTRICAS COMPLETAS")
print("="*100)

# =====================================================================
# PASO 1: CARGAR Y SIMPLIFICAR DATOS
# =====================================================================
print("\n PASO 1: Cargando datos...")
train_data = spark.read.parquet("../data/train_data.parquet")
test_data = spark.read.parquet("../data/test_data.parquet")

# Simplificar: Solo Tarjeta (1) vs Efectivo (2)
train = train_data.filter(col("payment_type").isin([1, 2]))
test = test_data.filter(col("payment_type").isin([1, 2]))

print(f"   Train original: {train.count():,} registros")
print(f"   Test original:  {test.count():,} registros")

# =====================================================================
# PASO 2: UNDERSAMPLING MODERADO (Ratio 3:1)
# =====================================================================
print("\n PASO 2: Aplicando undersampling...")
class_1 = train.filter(col("payment_type") == 1)
class_2 = train.filter(col("payment_type") == 2)

class_1_count = class_1.count()
class_2_count = class_2.count()

print(f"   Clase 1 (Tarjeta):  {class_1_count:,}")
print(f"   Clase 2 (Efectivo): {class_2_count:,}")
print(f"   Ratio original: {class_1_count/class_2_count:.1f}:1")

target_size = class_2_count * 3
fraction = target_size / class_1_count

class_1_sampled = class_1.sample(False, fraction, seed=42)
train_balanced = class_1_sampled.union(class_2)

print(f"   Train balanceado: {train_balanced.count():,} registros")
print(f"   Nuevo ratio: 3:1")

# =====================================================================
# PASO 3: CLASS WEIGHTS
# =====================================================================
print("\n  PASO 3: Calculando class weights...")
total = train_balanced.count()
count_1 = train_balanced.filter(col("payment_type") == 1).count()
count_2 = train_balanced.filter(col("payment_type") == 2).count()

weight_1 = total / (2 * count_1)
weight_2 = total / (2 * count_2)

print(f"   Weight clase 1: {weight_1:.4f}")
print(f"   Weight clase 2: {weight_2:.4f}")

train_weighted = train_balanced.withColumn(
    "class_weight",
    when(col("payment_type") == 1, weight_1).otherwise(weight_2)
)

# =====================================================================
# PASO 4: PREPARAR FEATURES
# =====================================================================
print("\n PASO 4: Preparando features...")
feature_cols = ["trip_distance", "total_amount", "passenger_count", 
                "pickup_hour", "pickup_dayofweek", "trip_duration_minutes"]

assembler = VectorAssembler(inputCols=feature_cols, outputCol="features")
train_assembled = assembler.transform(train_weighted)
test_assembled = assembler.transform(test)

# =====================================================================
# PASO 5: INDEXAR LABELS
# =====================================================================
print("\n  PASO 5: Indexando labels...")
indexer = StringIndexer(inputCol="payment_type", outputCol="label")
indexer_model = indexer.fit(train_assembled)
train_final = indexer_model.transform(train_assembled)
test_final = indexer_model.transform(test_assembled)

# Verificar mapeo de labels
print("   Mapeo de labels:")
train_final.select("payment_type", "label").distinct().orderBy("label").show()

# =====================================================================
# PASO 6: ENTRENAR MODELO
# =====================================================================
print("\n PASO 6: Entrenando modelo...")
lr = LogisticRegression(
    featuresCol="features",
    labelCol="label",
    weightCol="class_weight",
    maxIter=20
)

model = lr.fit(train_final)
print("    Modelo entrenado!")

# =====================================================================
# PASO 7: HACER PREDICCIONES
# =====================================================================
print("\n PASO 7: Haciendo predicciones en TEST...")
predictions = model.transform(test_final)

# Cache para múltiples evaluaciones
predictions.cache()
print(f"   Predicciones generadas: {predictions.count():,} registros")

# =====================================================================
# PASO 8: CALCULAR MÉTRICAS COMPLETAS
# =====================================================================
print("\n" + "="*100)
print(" PASO 8: EVALUACIÓN DEL MODELO - MÉTRICAS COMPLETAS")
print("="*100)

# 8.1 - AUC-ROC
evaluator_auc = BinaryClassificationEvaluator(labelCol="label", metricName="areaUnderROC")
auc = evaluator_auc.evaluate(predictions)

# 8.2 - AUC-PR (Area Under Precision-Recall Curve)
evaluator_pr = BinaryClassificationEvaluator(labelCol="label", metricName="areaUnderPR")
auc_pr = evaluator_pr.evaluate(predictions)

# 8.3 - Accuracy (MulticlassClassificationEvaluator)
evaluator_accuracy = MulticlassClassificationEvaluator(labelCol="label", predictionCol="prediction", metricName="accuracy")
accuracy = evaluator_accuracy.evaluate(predictions)

# 8.4 - Precision Macro
evaluator_precision = MulticlassClassificationEvaluator(labelCol="label", predictionCol="prediction", metricName="weightedPrecision")
precision = evaluator_precision.evaluate(predictions)

# 8.5 - Recall Macro
evaluator_recall = MulticlassClassificationEvaluator(labelCol="label", predictionCol="prediction", metricName="weightedRecall")
recall = evaluator_recall.evaluate(predictions)

# 8.6 - F1-Score
evaluator_f1 = MulticlassClassificationEvaluator(labelCol="label", predictionCol="prediction", metricName="f1")
f1 = evaluator_f1.evaluate(predictions)

print(f"\n MÉTRICAS GLOBALES:")
print(f"   Accuracy:           {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"   Precision (Weighted): {precision:.4f}")
print(f"   Recall (Weighted):    {recall:.4f}")
print(f"   F1-Score:           {f1:.4f}")
print(f"   AUC-ROC:            {auc:.4f}")
print(f"   AUC-PR:             {auc_pr:.4f}")

# =====================================================================
# PASO 9: MATRIZ DE CONFUSIÓN DETALLADA
# =====================================================================
print("\n" + "="*100)
print(" MATRIZ DE CONFUSIÓN")
print("="*100)

# 9.1 - Obtener conteos de la matriz de confusión
confusion_matrix = predictions.groupBy("label", "prediction").count().orderBy("label", "prediction")

print("\n Tabla de confusión:")
confusion_matrix.show()

# 9.2 - Calcular TP, TN, FP, FN manualmente
# Nota: label 0.0 = payment_type 1 (Tarjeta), label 1.0 = payment_type 2 (Efectivo)
# Consideramos clase positiva = 1.0 (Efectivo)

TP = predictions.filter((col("label") == 1.0) & (col("prediction") == 1.0)).count()
TN = predictions.filter((col("label") == 0.0) & (col("prediction") == 0.0)).count()
FP = predictions.filter((col("label") == 0.0) & (col("prediction") == 1.0)).count()
FN = predictions.filter((col("label") == 1.0) & (col("prediction") == 0.0)).count()

total_predictions = TP + TN + FP + FN

print("\n MATRIZ DE CONFUSIÓN DETALLADA:")
print("   (Clase Positiva = payment_type 2 - Efectivo)")
print(f"""
                          Predicción
                    |  Tarjeta (0)  |  Efectivo (1)  |
    ----------------+---------------+----------------+
    Real            |               |                |
    Tarjeta (0)     |  TN: {TN:>7,}  |  FP: {FP:>7,}   |  {TN+FP:,}
                    | ({TN/total_predictions*100:>5.2f}%)   | ({FP/total_predictions*100:>5.2f}%)    |
    ----------------+---------------+----------------+
    Efectivo (1)    |  FN: {FN:>7,}  |  TP: {TP:>7,}   |  {FN+TP:,}
                    | ({FN/total_predictions*100:>5.2f}%)   | ({TP/total_predictions*100:>5.2f}%)    |
    ----------------+---------------+----------------+
                    |  {TN+FN:,}      |  {FP+TP:,}       |  {total_predictions:,}
""")

# 9.3 - Calcular métricas por clase
print("\n MÉTRICAS POR CLASE:")

# Para clase 1 (Efectivo) - Positiva
precision_1 = TP / (TP + FP) if (TP + FP) > 0 else 0
recall_1 = TP / (TP + FN) if (TP + FN) > 0 else 0
f1_1 = 2 * (precision_1 * recall_1) / (precision_1 + recall_1) if (precision_1 + recall_1) > 0 else 0

# Para clase 0 (Tarjeta) - Negativa
precision_0 = TN / (TN + FN) if (TN + FN) > 0 else 0
recall_0 = TN / (TN + FP) if (TN + FP) > 0 else 0
f1_0 = 2 * (precision_0 * recall_0) / (precision_0 + recall_0) if (precision_0 + recall_0) > 0 else 0

print(f"\n   CLASE 0 (Tarjeta - payment_type 1):")
print(f"      Precision: {precision_0:.4f}")
print(f"      Recall:    {recall_0:.4f}")
print(f"      F1-Score:  {f1_0:.4f}")
print(f"      Support:   {TN+FP:,}")

print(f"\n   CLASE 1 (Efectivo - payment_type 2):")
print(f"      Precision: {precision_1:.4f}")
print(f"      Recall:    {recall_1:.4f}")
print(f"      F1-Score:  {f1_1:.4f}")
print(f"      Support:   {TP+FN:,}")

# 9.4 - Accuracy manual (debe coincidir con la calculada antes)
accuracy_manual = (TP + TN) / total_predictions

print(f"\n VERIFICACIÓN:")
print(f"   Accuracy (calculada):   {accuracy:.4f}")
print(f"   Accuracy (manual):      {accuracy_manual:.4f}")
print(f"    Coinciden: {'Sí' if abs(accuracy - accuracy_manual) < 0.0001 else 'No'}")

# =====================================================================
# PASO 10: ANÁLISIS DE ERRORES
# =====================================================================
print("\n" + "="*100)
print(" ANÁLISIS DE ERRORES")
print("="*100)

# 10.1 - Falsos Positivos (predice Efectivo, pero es Tarjeta)
false_positives = predictions.filter((col("label") == 0.0) & (col("prediction") == 1.0))
print(f"\n FALSOS POSITIVOS (FP): {FP:,}")
print("   (Predice EFECTIVO cuando en realidad es TARJETA)")
if FP > 0:
    print("\n   Primeros 5 ejemplos de FP:")
    false_positives.select("trip_distance", "total_amount", "passenger_count", 
                          "pickup_hour", "payment_type", "label", "prediction", "probability") \
                   .show(5, truncate=False)

# 10.2 - Falsos Negativos (predice Tarjeta, pero es Efectivo)
false_negatives = predictions.filter((col("label") == 1.0) & (col("prediction") == 0.0))
print(f"\n FALSOS NEGATIVOS (FN): {FN:,}")
print("   (Predice TARJETA cuando en realidad es EFECTIVO)")
if FN > 0:
    print("\n   Primeros 5 ejemplos de FN:")
    false_negatives.select("trip_distance", "total_amount", "passenger_count", 
                          "pickup_hour", "payment_type", "label", "prediction", "probability") \
                   .show(5, truncate=False)

# =====================================================================
# PASO 11: RESUMEN FINAL
# =====================================================================
print("\n" + "="*100)
print(" RESUMEN FINAL")
print("="*100)

print(f"""
 MODELO ENTRENADO Y EVALUADO EXITOSAMENTE

 DATASET:
   Train original:     {train.count():>10,} registros
   Train balanceado:   {train_balanced.count():>10,} registros
   Test:               {test.count():>10,} registros

⚖  BALANCE APLICADO:
   Undersampling: Ratio 3:1
   Class Weights: Sí (weight_1={weight_1:.4f}, weight_2={weight_2:.4f})

 RESULTADOS:
   Accuracy:     {accuracy:.4f} ({accuracy*100:.2f}%)
   Precision:    {precision:.4f}
   Recall:       {recall:.4f}
   F1-Score:     {f1:.4f}
   AUC-ROC:      {auc:.4f}

 MATRIZ DE CONFUSIÓN:
   True Positives (TP):  {TP:>7,} ({TP/total_predictions*100:>5.2f}%)
   True Negatives (TN):  {TN:>7,} ({TN/total_predictions*100:>5.2f}%)
   False Positives (FP): {FP:>7,} ({FP/total_predictions*100:>5.2f}%)
   False Negatives (FN): {FN:>7,} ({FN/total_predictions*100:>5.2f}%)

 INTERPRETACIÓN:
   - El modelo clasifica correctamente {accuracy*100:.2f}% de los casos
   - Precision clase Efectivo: {precision_1:.2%} de predicciones de efectivo son correctas
   - Recall clase Efectivo: {recall_1:.2%} de casos reales de efectivo son detectados
""")

print("="*100)
print(" EVALUACIÓN COMPLETA FINALIZADA")
print("="*100)

# Limpiar cache
predictions.unpersist()
````
