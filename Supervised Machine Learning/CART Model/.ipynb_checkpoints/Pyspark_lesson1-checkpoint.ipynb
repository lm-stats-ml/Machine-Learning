```python
#  código para hacer la regresión logística:
```

```python
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.classification import LogisticRegression
from pyspark.ml.evaluation import BinaryClassificationEvaluator

# 1. Preparar los datos - eliminar nulos
df_modelo = df_secop.filter(
    col('tiene_sobrecosto').isNotNull() &
    col('es_prestacion_servicios').isNotNull() &
    col('es_suministros').isNotNull() &
    col('es_compraventa').isNotNull() &
    col('es_obra').isNotNull() &
    col('es_arrendamiento_inmuebles').isNotNull() &
    col('es_interventoria').isNotNull() &
    col('es_consultoria').isNotNull() &
    col('es_seguros').isNotNull() &
    col('es_arrendamiento_muebles').isNotNull()
)

print(f"✅ Datos preparados: {df_modelo.count():,} registros")
```

```python
# 2. Crear el vector de features (variables independientes)
feature_cols = [
    'es_prestacion_servicios',
    'es_suministros',
    'es_compraventa',
    'es_obra',
    'es_arrendamiento_inmuebles',
    'es_interventoria',
    'es_consultoria',
    'es_seguros',
    'es_arrendamiento_muebles'
]

assembler = VectorAssembler(
    inputCols=feature_cols,
    outputCol='features'
)

df_features = assembler.transform(df_modelo)

print("✅ Vector de features creado")
```

```python
# 3. Dividir en train y test
train_data, test_data = df_features.randomSplit([0.8, 0.2], seed=42)

print(f"Datos de entrenamiento: {train_data.count():,}")
print(f"Datos de prueba: {test_data.count():,}")
```

```python
# 4. Entrenar el modelo de regresión logística
lr = LogisticRegression(
    featuresCol='features',
    labelCol='tiene_sobrecosto',
    maxIter=100
)

modelo = lr.fit(train_data)

print("✅ Modelo entrenado")
```

```python
# 5. Ver los coeficientes del modelo
import pandas as pd

coeficientes = modelo.coefficients.toArray()
intercepto = modelo.intercept

# Crear DataFrame con los resultados
resultados = pd.DataFrame({
    'Variable': feature_cols,
    'Coeficiente': coeficientes
})

resultados = resultados.sort_values('Coeficiente', ascending=False)

print("\n📊 COEFICIENTES DEL MODELO:")
print("="*60)
print(f"Intercepto: {intercepto:.4f}\n")
print(resultados.to_string(index=False))
```

```python
# 6. Hacer predicciones en datos de prueba
predicciones = modelo.transform(test_data)

# Ver algunas predicciones
predicciones.select(
    'tiene_sobrecosto',
    'prediction',
    'probability',
    'tipo_de_contrato'
).show(20)
```

```python
# 7. Evaluar el modelo
evaluator = BinaryClassificationEvaluator(
    labelCol='tiene_sobrecosto',
    rawPredictionCol='rawPrediction',
    metricName='areaUnderROC'
)

auc = evaluator.evaluate(predicciones)
print(f"\n📈 Área bajo la curva ROC (AUC): {auc:.4f}")
```

```python
# 8. Matriz de confusión
from pyspark.sql.functions import sum as spark_sum

confusion = predicciones.groupBy('tiene_sobrecosto', 'prediction').count()
confusion.orderBy('tiene_sobrecosto', 'prediction').show()
```

```python
# 9. Calcular precisión (accuracy)
total = predicciones.count()
correctos = predicciones.filter(col('tiene_sobrecosto') == col('prediction')).count()
precision = (correctos / total) * 100

print(f"\n✅ Precisión del modelo: {precision:.2f}%")
```

```python
# 10. Interpretación de coeficientes (Odds Ratios)
import numpy as np

resultados['Odds_Ratio'] = np.exp(resultados['Coeficiente'])
resultados['Incremento_%'] = (resultados['Odds_Ratio'] - 1) * 100

print("\n📊 ODDS RATIOS E INTERPRETACIÓN:")
print("="*80)
print(resultados.to_string(index=False))
```

Para obtener los valores p y la significancia estadística, necesitamos usar un enfoque diferente. PySpark MLlib no proporciona valores p directamente, pero podemos usar statsmodels con una muestra de datos.

Opción 1: Usar statsmodels con una muestra (Recomendado)

```python
# Convertir una muestra a Pandas para usar statsmodels
import pandas as pd
import statsmodels.api as sm

# Tomar una muestra manejable (ajusta el tamaño según tu memoria)
df_muestra = df_modelo.select(
    'tiene_sobrecosto',
    'es_prestacion_servicios',
    'es_suministros',
    'es_compraventa',
    'es_obra',
    'es_arrendamiento_inmuebles',
    'es_interventoria',
    'es_consultoria',
    'es_seguros',
    'es_arrendamiento_muebles'
).sample(fraction=0.1, seed=42).toPandas()  # 10% de los datos

print(f"Muestra para análisis: {len(df_muestra):,} registros")
```

```python
# Preparar variables independientes y dependiente
X = df_muestra[[
    'es_prestacion_servicios',
    'es_suministros',
    'es_compraventa',
    'es_obra',
    'es_arrendamiento_inmuebles',
    'es_interventoria',
    'es_consultoria',
    'es_seguros',
    'es_arrendamiento_muebles'
]]

y = df_muestra['tiene_sobrecosto']

# Agregar constante (intercepto)
X = sm.add_constant(X)
```

```python
# Ajustar modelo de regresión logística
modelo_logit = sm.Logit(y, X).fit()

# Ver resumen completo con valores p
print(modelo_logit.summary())
```

```python
# Extraer solo coeficientes, valores p y odds ratios
import numpy as np

resultados_detallados = pd.DataFrame({
    'Variable': modelo_logit.params.index,
    'Coeficiente': modelo_logit.params.values,
    'Error_Std': modelo_logit.bse.values,
    'z_value': modelo_logit.tvalues.values,
    'Valor_p': modelo_logit.pvalues.values,
    'IC_2.5%': modelo_logit.conf_int()[0].values,
    'IC_97.5%': modelo_logit.conf_int()[1].values
})

resultados_detallados['Odds_Ratio'] = np.exp(resultados_detallados['Coeficiente'])
resultados_detallados['Significativo'] = resultados_detallados['Valor_p'] < 0.05

print("\n📊 RESULTADOS DETALLADOS DE LA REGRESIÓN LOGÍSTICA:")
print("="*100)
print(resultados_detallados.to_string(index=False))
```

Ahora continuar con el modelo usando datos balanceados

```python
from pyspark.sql.functions import col

# Contar cada clase
count_clase_0 = df_secop.filter(col('tiene_sobrecosto') == 0).count()
count_clase_1 = df_secop.filter(col('tiene_sobrecosto') == 1).count()

print(f"Clase 0 (sin sobrecosto): {count_clase_0:,}")
print(f"Clase 1 (con sobrecosto): {count_clase_1:,}")

# Separar clases
df_clase_0 = df_secop.filter(col('tiene_sobrecosto') == 0)
df_clase_1 = df_secop.filter(col('tiene_sobrecosto') == 1)

# Definir ratio deseado (1:2 -> por cada 1 con sobrecosto, 2 sin sobrecosto)
ratio_deseado = 2

# Calcular fracción de muestreo
fraction_clase_0 = (count_clase_1 * ratio_deseado) / count_clase_0

print(f"\nFracción de muestreo para clase 0: {fraction_clase_0:.6f}")

# Muestrear clase mayoritaria
df_clase_0_sampled = df_clase_0.sample(fraction=fraction_clase_0, seed=42)

# Combinar
df_balanceado = df_clase_0_sampled.union(df_clase_1)

print(f"\n✅ Dataset balanceado con ratio 1:{ratio_deseado}")
print(f"Total registros: {df_balanceado.count():,}")

# Verificar balance
print("\n📊 DISTRIBUCIÓN BALANCEADA:")
balance = df_balanceado.groupBy('tiene_sobrecosto').count().orderBy('tiene_sobrecosto')
balance.show()

# Calcular porcentajes del balance
total_balanceado = df_balanceado.count()
from pyspark.sql.functions import round as spark_round

balance_porcentaje = df_balanceado.groupBy('tiene_sobrecosto') \
    .count() \
    .withColumn('porcentaje', spark_round((col('count') / total_balanceado) * 100, 2)) \
    .orderBy('tiene_sobrecosto')

print("\n📊 BALANCE CON PORCENTAJES:")
balance_porcentaje.show()
```

```python
# Preparar datos balanceados
df_modelo_balanceado = df_balanceado.filter(
    col('tiene_sobrecosto').isNotNull() &
    col('es_prestacion_servicios').isNotNull() &
    col('tiene_recursos_propios').isNotNull() &
    col('tiene_recursos_credito').isNotNull() &
    col('es_suministros').isNotNull() &
    col('es_compraventa').isNotNull() &
    col('es_obra').isNotNull() &
    col('es_arrendamiento_inmuebles').isNotNull() &
    col('es_interventoria').isNotNull() &
    col('es_consultoria').isNotNull() &
    col('es_seguros').isNotNull() &
    col('es_postconflicto').isNotNull() &
    col('es_arrendamiento_muebles').isNotNull()
)

print(f"✅ Datos balanceados preparados: {df_modelo_balanceado.count():,} registros")
```

```python
# Crear features
from pyspark.ml.feature import VectorAssembler

feature_cols = [
    'es_prestacion_servicios',
    'tiene_recursos_propios',
    'tiene_recursos_credito',
    'es_suministros',
    'es_compraventa',
    'es_obra',
    'es_arrendamiento_inmuebles',
    'es_interventoria',
    'es_consultoria',
    'es_seguros',
    'es_arrendamiento_muebles',
     'es_postconflicto'
]

assembler = VectorAssembler(inputCols=feature_cols, outputCol='features')
df_features_balanceado = assembler.transform(df_modelo_balanceado)
```

```python
# Dividir en train y test
train_data, test_data = df_features_balanceado.randomSplit([0.7, 0.3], seed=42)

print(f"Datos de entrenamiento: {train_data.count():,}")
print(f"Datos de prueba: {test_data.count():,}")

# Verificar balance en train y test
print("\nBalance en Train:")
train_data.groupBy('tiene_sobrecosto').count().show()

print("Balance en Test:")
test_data.groupBy('tiene_sobrecosto').count().show()
```

```python
# 8. Para obtener valores p - usar statsmodels con muestra
import statsmodels.api as sm

# Convertir muestra a Pandas
df_pandas = df_modelo_balanceado.select(
    'tiene_sobrecosto',
    'es_prestacion_servicios',
    'tiene_recursos_propios',
    'es_suministros',
    'es_compraventa',
    'es_obra',
    'es_arrendamiento_inmuebles',
    'es_interventoria',
    'es_consultoria',
    'es_seguros',
    'es_arrendamiento_muebles'
).toPandas()

X = df_pandas[feature_cols]
y = df_pandas['tiene_sobrecosto']
X = sm.add_constant(X)

# Ajustar modelo
modelo_logit = sm.Logit(y, X).fit()

print("\n📊 RESUMEN COMPLETO CON VALORES P:")
print("="*80)
print(modelo_logit.summary())
```

```python
# Visualización resumida
print("\n📋 RESUMEN - TOP 5 CON MAYOR SOBRECOSTO:")
print("="*80)

# Por porcentaje
print("\n🔝 Por PORCENTAJE de contratos con sobrecosto:")
analisis_sobrecosto.select('tipo_de_contrato', 'porcentaje_sobrecosto') \
    .orderBy('porcentaje_sobrecosto', ascending=False) \
    .show(5, truncate=False)

# Por monto promedio
print("\n💰 Por MONTO PROMEDIO de sobrecosto:")
analisis_sobrecosto.select('tipo_de_contrato', 'sobrecosto_promedio') \
    .orderBy('sobrecosto_promedio', ascending=False) \
    .show(5, truncate=False)
```

```python
# 1. Hacer predicciones con el modelo
y_pred_prob = modelo_logit.predict(X)  # Probabilidades
y_pred = (y_pred_prob > 0.5).astype(int)  # Clasificación binaria (umbral 0.5)

print("✅ Predicciones calculadas")
```

```python
# 2. Matriz de confusión
from sklearn.metrics import confusion_matrix, classification_report
import numpy as np

conf_matrix = confusion_matrix(y, y_pred)

print("\n📊 MATRIZ DE CONFUSIÓN:")
print("="*60)
print(f"                  Predicho: 0    Predicho: 1")
print(f"Real: 0 (TN, FP)    {conf_matrix[0,0]:8d}    {conf_matrix[0,1]:8d}")
print(f"Real: 1 (FN, TP)    {conf_matrix[1,0]:8d}    {conf_matrix[1,1]:8d}")

TN, FP, FN, TP = conf_matrix.ravel()
print(f"\nTrue Negatives (TN):  {TN:,}")
print(f"False Positives (FP): {FP:,}")
print(f"False Negatives (FN): {FN:,}")
print(f"True Positives (TP):  {TP:,}")
```

```python
# 3. Métricas de desempeño
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

accuracy = accuracy_score(y, y_pred)
precision = precision_score(y, y_pred)
recall = recall_score(y, y_pred)
f1 = f1_score(y, y_pred)
auc_roc = roc_auc_score(y, y_pred_prob)

print("\n📈 MÉTRICAS DE DESEMPEÑO:")
print("="*60)
print(f"Accuracy (Exactitud):          {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"Precision (Precisión):         {precision:.4f} ({precision*100:.2f}%)")
print(f"Recall (Sensibilidad):         {recall:.4f} ({recall*100:.2f}%)")
print(f"F1-Score:                      {f1:.4f}")
print(f"AUC-ROC:                       {auc_roc:.4f}")
```
