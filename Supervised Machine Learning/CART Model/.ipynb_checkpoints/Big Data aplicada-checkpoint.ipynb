### Contexto: El Problema del Big Data


```
¿Por qué necesitamos un nuevo motor y modelo de programación para la analítica de datos en primer lugar? Como con muchas tendencias en la computación, esto se debe a cambios en los factores económicos que subyacen a las aplicaciones informáticas y el hardware.


Durante la mayor parte de su historia, las computadoras se volvían más rápidas cada año mediante aumentos en la velocidad del procesador: los nuevos procesadores cada año podían ejecutar más instrucciones por segundo que los del año anterior. Como resultado, las aplicaciones también se volvían automáticamente más rápidas cada año, sin necesidad de ningún cambio en su código. Esta tendencia llevó a un ecosistema grande y establecido de aplicaciones que se construyó con el tiempo, la mayoría de las cuales fueron diseñadas para ejecutarse solo en un único procesador. Estas aplicaciones aprovecharon la tendencia de mejora en las velocidades de procesador para escalar a cómputos más grandes y mayores volúmenes de datos con el tiempo.

Desafortunadamente, esta tendencia en el hardware se detuvo alrededor de 2005: debido a límites rígidos en la disipación de calor, los desarrolladores de hardware dejaron de hacer procesadores individuales más rápidos, y cambiaron hacia agregar más núcleos de CPU paralelos todos ejecutándose a la misma velocidad. Este cambio significó que de repente las aplicaciones necesitaban ser modificadas para agregar paralelismo con el fin de ejecutarse más rápido, lo que preparó el escenario para nuevos modelos de programación como Apache Spark.

Además de eso, las tecnologías para almacenar y recolectar datos no se desaceleraron apreciablemente en 2005, cuando las velocidades de procesador lo hicieron. El costo de almacenar 1 TB de datos continúa cayendo aproximadamente dos veces cada 14 meses, lo que significa que es muy económico para organizaciones de todos los tamaños almacenar grandes cantidades de datos. Además, muchas de las tecnologías para recolectar datos (sensores, cámaras, conjuntos de datos públicos, etc.) continúan bajando de costo y mejorando en resolución. Por ejemplo, la tecnología de cámaras continúa mejorando en resolución y bajando en costo por píxel cada año, hasta el punto donde una cámara web de 12 megapíxeles cuesta solo $3 a $4; esto ha hecho económico recolectar una amplia gama de datos visuales, ya sea de personas filmando video o sensores automatizados en un entorno industrial. Además, las cámaras son en sí mismas los sensores clave en otros dispositivos de recolección de datos, como telescopios e incluso máquinas de secuenciación genética, impulsando también la baja de costos de estas tecnologías
.
El resultado final es un mundo en el cual recolectar datos es extremadamente económico—muchas organizaciones hoy en día incluso consideran negligente no registrar datos de posible relevancia para el negocio—pero procesarlos requiere cómputos grandes y paralelos, a menudo en clústeres de máquinas. Además, en este nuevo mundo, el software desarrollado en los últimos 50 años no puede escalar automáticamente, y tampoco pueden hacerlo los modelos de programación tradicionales para aplicaciones de procesamiento de datos, creando la necesidad de nuevos modelos de programación. Es para este mundo que Apache Spark fue construido.

### Que es Apache Spark?

```
Apache Spark es un motor de computación unificado y un conjunto de bibliotecas para el procesamiento paralelo de datos en clústeres de computadoras. Spark soporta múltiples lenguajes de programación ampliamente utilizados (Python, Java, Scala y R), incluye bibliotecas para diversas tareas que van desde SQL hasta streaming y aprendizaje automático, y se ejecuta en cualquier lugar, desde una computadora portátil hasta un clúster de miles de servidores. Esto lo convierte en un sistema fácil de comenzar a usar y de escalar hasta el procesamiento de big data o a una escala increíblemente grande. 



### Bibliotecas

```
El componente final de Spark son sus bibliotecas, que se construyen sobre su diseño como motor unificado para proporcionar una API unificada para tareas comunes de análisis de datos. Spark soporta tanto bibliotecas estándar que vienen con el motor como una amplia variedad de bibliotecas externas publicadas como paquetes de terceros por las comunidades de código abierto. Hoy en día, las bibliotecas estándar de Spark son en realidad la mayor parte del proyecto de código abierto: el motor central de Spark en sí mismo ha cambiado poco desde que fue lanzado por primera vez, pero las bibliotecas han crecido para proporcionar cada vez más tipos de funcionalidad.

Spark incluye bibliotecas para SQL y datos estructurados (Spark SQL), aprendizaje automático (MLlib), procesamiento de streaming (Spark Streaming y el más reciente Structured Streaming), y análisis de grafos (GraphX). Más allá de estas bibliotecas, existen cientos de bibliotecas externas de código abierto que van desde conectores para varios sistemas de almacenamiento hasta algoritmos de aprendizaje automático. Un índice de bibliotecas externas está disponible en spark-packages.org

### Spark vs Pandas en datos masivos

```

La diferencia fundamental está en cómo procesan los datos:

Pandas trabaja en una sola máquina, en memoria RAM. Si tienes 10 GB de datos y tu computador tiene 16 GB de RAM, Pandas intenta cargar todo ahí. Cada operación (filtrar, agrupar, entrenar un modelo) se ejecuta de forma secuencial en un solo núcleo del procesador. Si los datos no caben en memoria, simplemente falla. Spark distribuye los datos en particiones a través de múltiples núcleos o múltiples máquinas. Un DataFrame de 10 GB se divide, por ejemplo, en 100 particiones de 100 MB cada una, y las operaciones se ejecutan en paralelo sobre todas las particiones simultáneamente.

Pensalo así con una analogía simple: si tienes que contar 4 millones de registros de SECOP, Pandas es una sola persona contando uno por uno. Spark es un equipo de 8 o 16 personas, cada una contando su porción al mismo tiempo.


Hay otras razones técnicas importantes. Spark usa evaluación perezosa (lazy evaluation): cuando escribís
 
NYtaxidata2024.where("trip_distance > 10").groupBy("PULocationID").sum("total_amount"),  Spark no ejecuta nada inmediatamente. Primero construye un plan lógico completo, lo optimiza con su motor Catalyst (reordena operaciones, elimina pasos innecesarios), y solo ejecuta cuando llamás .show() o .collect(). Pandas en cambio ejecuta cada línea inmediatamente, sin optimización global.

### Iniciando Spark
```
Hasta ahora, hemos cubierto los conceptos básicos de las Aplicaciones Spark. Todo esto ha sido de naturaleza conceptual. Cuando realmente vayamos a escribir nuestra Aplicación Spark, vamos a necesitar una forma de enviar comandos de usuario y datos a ella. Hacemos eso primero creando un SparkSession.


````python
# Crear sesión 
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Spark_Applied Course") \
    .master("local[*]") \
    .config("spark.driver.memory", "4g") \
    .config("spark.executor.memory", "4g") \
    .getOrCreate()

print("Spark iniciado correctamente")
````

#### Explicacion

```

- from pyspark.sql import SparkSession: Importa la clase SparkSession, que es el punto de entrada principal a Spark cuando trabajas con DataFrames y SQL.

- spark = SparkSession.builder \ : Llama al constructor de configuración de la sesión de Spark.

- appName("Spark_Applied Course") \Define el nombre de la aplicación Spark. Este nombre aparece en:
•	Spark UI
•	logs
    
- .master("local[*]"): \Indica dónde se ejecutará Spark "local[*]" significa:
•	Ejecutar en modo local
•	Usar todos los núcleos del CPU disponibles

- config("spark.driver.memory", "4g") \

•  Asigna 4 GB de memoria a cada executor.
•  Los executors:
•	Ejecutan las tareas distribuidas
•	Procesan los datos realmente
En modo local[*], driver y executor suelen compartir recursos, pero la configuración sigue siendo válida.

- .getOrCreate():Hace dos cosas:
•  Crea una nueva SparkSession si no existe
•  Reutiliza una existente si ya está activa
•  Evita errores por múltiples sesiones abiertas.

    El resultado se guarda en la variable spark.

### RDD
```
Los RDD (Resilient Distributed Datasets) son la **capa más baja** de Spark, los bloques fundamentales. Son colecciones de datos distribuidas en particiones a través del clúster.
Es decir, todo DataFrame y toda consulta SQL eventualmente se convierte en operaciones sobre RDDs. Catalyst es el intermediario inteligente que traduce tus instrucciones de alto nivel en operaciones RDD optimizadas.

¿Qué es Catalyst?
Catalyst es el optimizador de consultas interno de Spark. Su trabajo es tomar el código que vos escribís (ya sea SQL o DataFrame API) y transformarlo en el plan de ejecución más eficiente posible antes de ejecutarlo.

#### Inferschema. Nota.
```
Parquet ya trae el esquema embebido → inferSchema no aplica

# Sin inferSchema → todo queda como string
data = spark.read.csv("archivo.csv", header=True)

# Con inferSchema → Spark detecta int, double, timestamp, etc.
data = spark.read.csv("archivo.csv", header=True, inferSchema=True)

¿Cómo funciona internamente?
Spark hace dos pasadas sobre el archivo:

Primera pasada → lee los datos para inferir los tipos
Segunda pasada → carga los datos con los tipos detectados

Por eso tiene un costo: tarda el doble en archivos grandes.

```
Para añadir un poco más de especificidad a estas definiciones, necesitamos hablar de esquemas (schemas), que son la forma en que se definen los tipos de datos que se están almacenando en esta colección distribuida.

#### Qué es un schema?
```
Un schema es la estructura formal de un DataFrame.
Define:
•	Nombres de las columnas
•	Tipos de datos de cada columna
•	(En algunos sistemas) si una columna puede ser nula (nullable)
Es el equivalente a decir:
“Esta tabla tiene estas columnas y cada una guarda este tipo de información”.

Por qué es importante el schema?
Porque en sistemas distribuidos (como Spark):
•	Los datos están partidos en muchos nodos
•	Cada nodo debe saber qué tipo de dato está procesando
•	Las operaciones (filtros, joins, agregaciones) dependen del ti
Sin schema, Spark no sabría:
•	Si puede sumar una columna
•	Si puede comparar valores
•	Cómo optimizar la ejecucpType
n
po




```
“Schemas consist of types”
Esto significa que cada columna tiene un tipo explícito, por ejemplo:
•	IntegerType
•	DoubleType
•	StringType
•	DateType
•	TimestampType



````python
from pyspark.sql import SparkSession
from pyspark.sql.types import (
    StructType, StructField,
    IntegerType, DoubleType,
    StringType, DateType, TimestampType
)
from datetime import date, datetime

spark = SparkSession.builder.appName("SchemaExample").getOrCreate()

# Definimos el schema con tipos explícitos
schema = StructType([
    StructField("id", IntegerType(), True),
    StructField("nombre", StringType(), True),
    StructField("salario", DoubleType(), True),
    StructField("fecha_ingreso", DateType(), True),
    StructField("ultima_actualizacion", TimestampType(), True)
])

# Datos de ejemplo
data = [
    (1, "Ana", 3500.75, date(2022, 5, 10), datetime(2024, 1, 15, 10, 30, 0)),
    (2, "Luis", 4200.00, date(2021, 8, 1), datetime(2024, 1, 16, 9, 45, 0))
]

# Creamos el DataFrame usando el schema
df = spark.createDataFrame(data, schema=schema)

df.show(truncate=False)
df.printSchema()
````

### Columnas y expresiones
```
Las columnas en Spark son similares a las columnas en una hoja de cálculo, un dataframe de R o un DataFrame de pandas. Puedes seleccionar, manipular y eliminar columnas 
de los DataFrames, y estas operaciones se representan como expresiones.
Para Spark, las columnas son construcciones lógicas que simplemente representan un valor calculado para cada registro mediante una expresión. Esto significa que, para tener un valor real en una columna, necesitamos una fila; y para tener una fila, necesitamos un DataFrame. No es posible manipular una columna individual fuera del contexto de un DataFrame; debes utilizar transformaciones de Spark dentro de un DataFrame para modificar el contenido de una columna.

```
Como se mencionó en “Columnas y Expresiones”, puedes hacer referencia a las columnas de varias formas diferentes; lo único que debes tener en cuenta es que puedes usarlas de manera intercambiable:

````python
from pyspark.sql.functions import expr, col, column

NYtaxidata2024.select(
    expr("fare_amount"),
    col("fare_amount"),
    column("fare_amount"))\
.show(2)
````

```
expr es la referencia más flexible que podemos usar. Puede hacer referencia a una columna simple o a una manipulación de la columna expresada como cadena. Para ilustrarlo, cambiemos el nombre de una columna y luego volvamos a su nombre original usando la palabra clave AS y luego el método alias sobre la columna:

````python
# Renombrar columnas con alias
NYtaxidata2024.select(expr(
    "PULocationID as zona_recogida"
)).show(3)
````

```
Dado que select seguido de una serie de expr es un patrón tan común, Spark proporciona un atajo para hacerlo de manera más eficiente: selectExpr. Probablemente, esta sea la interfaz más conveniente para el uso diario.

#### Explicación:
```
`selectExpr()` permite usar **expresiones SQL** directamente como strings, incluyendo alias con `as`.
Esto abre el verdadero poder de Spark. Podemos tratar selectExpr como una forma sencilla de construir expresiones complejas que crean nuevos DataFrames. De hecho, podemos agregar cualquier sentencia SQL válida que no sea de agregación y, siempre que las columnas se resuelvan correctamente, será válida.



````python
# Con selectExpr
NYtaxidata2024.selectExpr(
    "((tip_amount / fare_amount) > 0.20) as buena_propina"
).show(5)
````

````python
# Con SQL
NYtaxidata2024.createOrReplaceTempView("taxi")
spark.sql("""
    SELECT ((tip_amount / fare_amount) > 0.20) as buena_propina
    FROM taxi
    LIMIT 5
""").show()
````

```
Este código crea una **nueva columna booleana** que indica si la propina es buena.

#### Divisiones aleatorias 
```
Las divisiones aleatorias pueden ser útiles cuando necesitas dividir tu DataFrame original en “particiones” aleatorias. Esto se utiliza con frecuencia en algoritmos de aprendizaje automático para crear conjuntos de entrenamiento, validación y prueba.
En el siguiente ejemplo, dividiremos nuestro DataFrame en dos DataFrames diferentes estableciendo los pesos con los que se realizará la división (estos son los argumentos de la función). Dado que este método está diseñado para ser aleatorio, también especificaremos una semilla (seed) para garantizar la reproducibilidad (solo reemplaza seed por un número de tu elección en el bloque de código).
Es importante tener en cuenta que, si no especificas proporciones para cada DataFrame que sumen uno, Spark las normalizará automáticamente para que así sea.

````python
seed = 42
dataFrames = data.randomSplit([0.25, 0.75], seed)

# Verificar la división
print(f"DataFrame 1 (25%): {dataFrames[0].count()} filas")
print(f"DataFrame 2 (75%): {dataFrames[1].count()} filas")
````

#### Ordenamiento de filas 
```
Cuando ordenamos los valores en un DataFrame, normalmente queremos que los valores más grandes o más pequeñosaparezcan en la parte superior del DataFrame. Existen dos operaciones equivalentes para realizar este ordenamiento: sorty orderBy, las cuales funcionan exactamente de la misma manera.
Ambas aceptan expresiones de columnas y cadenas de texto, así como múltiples columnas. De forma predeterminada, el ordenamiento se realiza en orden ascendente.



````python
# in Python
data.sort("fare_amount").show(5)
data.orderBy("fare_amount").show(5)
````

```
Para especificar de forma más explícita la dirección del ordenamiento, es necesario utilizar las funciones asc y desccuando se opera sobre una columna. Estas funciones permiten indicar el orden en el que debe ordenarse una columna determinada.

````python
# in Python
from pyspark.sql.functions import desc, asc

data.orderBy(col("fare_amount").desc()).show(5)
````

```
Un consejo avanzado es utilizar asc_nulls_first, desc_nulls_first, asc_nulls_last o desc_nulls_last para especificar en qué posición deseas que aparezcan los valores nulos dentro de un DataFrame ordenado.

#### Trabajo con números
```
Al trabajar con big data, la segunda tarea más común después de filtrar datos es contar cosas. En la mayoría de los casos, simplemente necesitamos expresar nuestro cálculo, y este será válido siempre que estemos trabajando con tipos de datos numéricos.

Imaginemos que necesitamos identificar el punto de inflexión a partir del cual por encima de una tarifa pagada disminuye la probabildad de dar propina. Para ello, para identificar rendimientos
decrecientes en una variable, debe ingresarse la variable al cuadrado dentro de un modelo probabilistico como un modelo logit (machine learning supervisado)

````python
from pyspark.sql.functions import col, pow 
data = data.withColumn("fare_amount2", pow(col("fare_amount"), 2))
data.select("fare_amount", "fare_amount2").show(5)
````

### Trabajo con fechas y marcas de tiempo
```

Las fechas y las horas son un desafío constante en los lenguajes de programación y en las bases de datos. Siempre es necesario llevar un control de las zonas horarias y asegurarse de que los formatos sean correctos y válidos. Spark hace todo lo posible por simplificar esto centrándose explícitamente en dos tipos de información temporal. Por un lado, están las fechas, que se enfocan exclusivamente en las fechas del calendario, y por otro, las marcas de tiempo (timestamps), que incluyen tanto la fecha como la hora.

````python
data.select("tpep_pickup_datetime", "tpep_dropoff_datetime").show(5, truncate=False)
data.printSchema()  # ya vimos que son timestamp_ntz (sin timezone)
````

````python
data = data.drop("integer")
````

````python
from pyspark.sql.functions import (
    year, month, dayofmonth, dayofweek,
    hour, round, unix_timestamp, when, col
)

# ── 1. Extraer componentes temporales ────────────────────────────────────────
data = data.withColumn("anio",          year("tpep_pickup_datetime")) \
           .withColumn("mes",           month("tpep_pickup_datetime")) \
           .withColumn("dia",           dayofmonth("tpep_pickup_datetime")) \
           .withColumn("dia_semana",    dayofweek("tpep_pickup_datetime")) \
           .withColumn("hora_recogida", hour("tpep_pickup_datetime"))

# ── 2. Duración del viaje en minutos ─────────────────────────────────────────
data = data.withColumn(
    "duracion_minutos",
    round((unix_timestamp("tpep_dropoff_datetime") -
           unix_timestamp("tpep_pickup_datetime")) / 60, 2)
)

# ── 3. Franja horaria ────────────────────────────────────────────────────────
data = data.withColumn("franja_horaria",
    when((col("hora_recogida") >= 6)  & (col("hora_recogida") < 10), "mañana_pico")
   .when((col("hora_recogida") >= 10) & (col("hora_recogida") < 16), "mediodia")
   .when((col("hora_recogida") >= 16) & (col("hora_recogida") < 20), "tarde_pico")
   .otherwise("nocturno")
)

# ── 4. Verificar resultados ──────────────────────────────────────────────────
data.select(
    "tpep_pickup_datetime",
    "hora_recogida", "dia_semana",
    "duracion_minutos", "franja_horaria"
).show(5, truncate=False)
````

```
Otra tarea común es analizar la diferencia entre dos fechas. Podemos hacerlo utilizando la función datediff, que devuelve el número de días entre dos fechas. En la mayoría de los casos solo nos interesan los días y, dado que la cantidad de días varía de un mes a otro, también existe la función months_between, que devuelve el número de meses entre dos fechas:

````python
from pyspark.sql.functions import avg, round, count, when, col

# ── 1. Añadir etiqueta legible al día de semana ───────────────────────────────
data = data.withColumn("nombre_dia",
    when(col("dia_semana") == 1, "Domingo")
   .when(col("dia_semana") == 2, "Lunes")
   .when(col("dia_semana") == 3, "Martes")
   .when(col("dia_semana") == 4, "Miercoles")
   .when(col("dia_semana") == 5, "Jueves")
   .when(col("dia_semana") == 6, "Viernes")
   .when(col("dia_semana") == 7, "Sabado")
)

# ── 2. Análisis por día de semana (resumen general) ───────────────────────────
print("=== COMPORTAMIENTO POR DÍA DE SEMANA ===")
data.groupBy("dia_semana", "nombre_dia").agg(
    round(avg("fare_amount"),      2).alias("tarifa_promedio"),
    round(avg("trip_distance"),    2).alias("distancia_promedio"),
    round(avg("duracion_minutos"), 2).alias("duracion_promedio"),
    round(avg("tip_amount"),       2).alias("propina_promedio"),
    count("*")                       .alias("total_viajes")
).orderBy("dia_semana").show()

# ── 3. Análisis por franja horaria y día (corregido) ──────────────────────────
print("=== COMPORTAMIENTO POR FRANJA HORARIA Y DÍA ===")
data.groupBy("dia_semana", "nombre_dia", "franja_horaria").agg(  # ← añadido dia_semana
    round(avg("fare_amount"),      2).alias("tarifa_promedio"),
    round(avg("duracion_minutos"), 2).alias("duracion_promedio"),
    count("*")                       .alias("total_viajes")
).orderBy("dia_semana", "franja_horaria").show(28)  # ← ahora sí existe dia_semana

# ── 4. Top horas con más viajes (hora pico real) ──────────────────────────────
print("=== TOP HORAS CON MÁS DEMANDA ===")
data.groupBy("hora_recogida").agg(
    count("*")                       .alias("total_viajes"),
    round(avg("fare_amount"),      2).alias("tarifa_promedio"),
    round(avg("duracion_minutos"), 2).alias("duracion_promedio")
).orderBy(col("total_viajes").desc()).show(10)
````

#### Valores no nulos

````python
from pyspark.sql.functions import col, sum, isnan, when
from pyspark.sql.types import DoubleType, FloatType, IntegerType, LongType

# Identificar columnas numéricas y no numéricas
tipos_numericos = (DoubleType, FloatType, IntegerType, LongType)

na_counts = data.select([
    sum(
        when(
            col(c).isNull() | (isnan(c) if isinstance(data.schema[c].dataType, tipos_numericos) else col(c).isNull()),
            1
        ).otherwise(0)
    ).alias(c)
    for c in data.columns
]).collect()[0].asDict()

print("Columnas con valores nulos:")
for columna, cantidad in na_counts.items():
    if cantidad > 0:
        print(f"  {columna}: {cantidad:,} NAs")
    else:
        print(f"  {columna}: OK")
````

### Particiones
```

Para permitir que cada executor realice trabajo en paralelo, Spark divide los datos en fragmentos llamados particiones. Una partición es una colección de filas que residen en una máquina física en tu clúster. Las particiones de un DataFrame representan cómo los datos están físicamente distribuidos a través del clúster de máquinas durante la ejecución

El número de particiones NO depende del número de archivos, sino de cómo Spark decide paralelizar el trabajo.

Una cosa importante a notar es que con los DataFrames no manipulas (en su mayor parte) las particiones manualmente o individualmente. Simplemente especificas transformaciones de alto nivel de los datos en las particiones físicas, y Spark determina cómo este trabajo será realmente ejecutado en el clúster. Existen APIs de nivel más bajo (a través de la interfaz RDD)

### Explicacion
#### Paso 0.Inicio: Lee el archivo. FileScan parquet (lectura)
```

FileScan parquet [fare_amount#10] Batched: true, DataFilters: [], Format: Parquet, Location: InMemoryFileIndex(1 paths)[file:/home/jovyan/data/Database/nyc_taxi_data/yellow_tripdata_2024-01...., PartitionFilters: [], PushedFilters: [], ReadSchema: struct<fare_amount:double>

Significa:
•	Spark lee solo la columna necesaria (fare_amount)
•	No lee todo el archivo Parquet (optimización columnar )
•	Lee en paralelo (múltiples partitions) eficien

Batched: true significa que Spark lee el archivo Parquet en modo vectorizado (columnar), procesando un bloque de muchas filas a la vez en lugar de fila por fila.
En términos concretos, cuando Spark lee tu archivo Parquet con Batched: true, internamente crea objetos ColumnVector que contienen miles de valores de fare_amount empaquetados juntos en memoria. El motor de ejecución vectorizado de Spark (Tungsten) puede entonces aplicar operaciones sobre ese bloque completo usando instrucciones SIMD del procesador, que operan sobre múltiples valores en un solo ciclo de CPU.

Esto es posible precisamente porque Parquet es un formato columnar: los valores de fare_amount ya están almacenados contiguamente en disco, lo que hace que la lectura vectorizada sea natural. Si el formato fuera CSV o JSON (formatos orientados a filas), Batched sería false porque los datos no están organizados por columna.
Para tu dataset de 4.4 millones de registros, esto se traduce en aproximadamente 1000 lotes de 4096 filas. La diferencia de rendimiento frente a lectura fila por fila puede ser de 2x a 5x solo en la fase de lectura.

El resto de los campos del FileScan también tienen su lógica: DataFilters: [] indica que no hay filtros que Spark pueda empujar dentro del lector Parquet (porque no hay un .filter() antes del groupBy), PushedFilters: [] confirma lo mismo, PartitionFilters: [] indica que el archivo no está particionado por directorios, y ReadSchema: struct<fare_amount:double> confirma que gracias al column pruning Spark solo leyó esa columna de todas las disponibles en el archivo.te


#### Paso 1.Selecciona columna. Project (renombrar columna)
```

 Project [fare_amount#10 AS fare#3113]

Aquí Spark:
•	Renombra fare_amount → fare (Es como un SELECT fare_amount AS fare.
•	No mueve datos
•	Es una transformación narr

Estado de los datos:

Partition 1: [fare=10.5, fare=20.0, fare=10.5, ...]
Partition 2: [fare=20.0, fare=10.5, fare=30.0, ...]
Partition 3: [fare=15.0, fare=10.5, fare=20.0, ]

fare_amount#10 es el ID que Spark le asignó a la columna fare_amount cuando leyó el archivo Parquet original. El 10 simplemente significa que fue la décima expresión/columna registrada en el plan lógico de esa sesión de Spark (ver en schema que ocupa la posición 10). No tiene relación con la posición de la columna en el archivo ni con ningún valor de los datos.
fare#3113 es el nuevo ID que Spark asignó a esa misma columna después de renombrarla con .withColumnRenamed("fare_amount", "fare"). El 3113 es mucho más alto porque Spark había registrado 3112 expresiones antes en esa sesión, contando todas las transformaciones intermedias, metadatos internos, expresiones del optimizador, y cualquier otra operación que haya ocurrido desde que arrancó el SparkContext.

La razón por la que Spark necesita estos IDs en lugar de solo usar el nombre de la columna es precisamente para evitar colisiones. Imagina que tienes dos columnas llamadas fare en un join de dos DataFrames, o que renombras una columna varias veces en cadena. Los nombres pueden repetirse, pero el ID numérico es siempre único en toda la sesión, lo que permite al optimizador saber exactamente de qué columna habla en cada nodo del árbol del plan, de dónde viene y cómo fue transformada.
Si reinicias la sesión de Spark y ejecutas el mismo código, los números cambiarán, pero el comportamiento del plan será idéntico. Por eso al depurar planes con explain() lo importante es entender la operación (Project, HashAggregate, etc.), no los números en sí..]

#### Paso 2.Normalización de valores flotantes (primero se hace esto y antes de hacer  Hashaggreate parcial o conteo parcial en el paso 4)
```
HashAggregate(keys=[knownfloatingpointnormalized(normalizenanandzero(fare#3113)) AS fare#3113], functions=[partial_count(1)])

Spark:
•	Trata NaN, -0.0, 0.0
•	Normaliza floats para que:
o	0.0 y -0.0 no se cuenten como grupos distintos
o	NaN no rompa el hash
Esto es automático, tú no lo pediste.

El paso de normalización es importante para tu dataset de taxis NYC porque los valores NaN o -0.0 en fare_amount son frecuentes, y sin esta normalización el groupBy produciría grupos duplicados para valores que deberían ser idénticos.

                                                        
Luego de esto:

Spark empieza a:
                                                                     
•	Construir estructuras hash locales
•	Preparar los datos antes del shuffle
Todavía:
•	no hay conteo
•	no hay redistribución

Qué hace?
•	Cada partition agrupa sus propios datos localmente
•	NO cruza partitions todavía
•	Es la fase de "map" en MapRed

Estado de los datos:
Partition 1 (local):
  {fare=10.5: 50 registros, fare=20.0: 30 registros}
Partition 2 (local):
  {fare=10.5: 25 registros, fare=20.0: 40 registros}
Partition 3 (local):
  {fare=10.5: 35 registros, fare=15.0: 20 registros}
IMPORTANTE:
•	Los datos aún NO están agrupados globalmente
•	Cada partition solo sabe sus propios conteos

#### Paso 3: Exchange/Shuffle (REORGANIZACIÓN - Paso costoso)
```

+- Exchange hashpartitioning(fare#3113, 200), ENSURE_REQUIREMENTS, [plan_id=528]
Este es el corazón del groupBy

Significa:
•	Spark redistribuye los datos
•	Todas las filas con el mismo fare
→ van a la misma partición
•	200 = número de particiones por defecto (spark.sql.shuffle.partitions)
Aquí:
•	hay red
•	hay disco
•	hay costo
Esto es lo que hace al groupBy una wide trans¿Qué hace?
•	REORGANIZA todos los datos entre partitions
•	Usa hash para decidir a qué partition va cada fare
•	Crea 200 nuevas partitions (por defecto)
¿Por qué es necesario? Porque antes del shuffle:
Partition 1: {fare=10.5: 50, fare=20.0: 30}
Partition 2: {fare=10.5: 25, fare=20.0: 40}
Partition 3: {fare=10.5: 35, fare=15.0: 20}
Todos los fare=10.5 están en DIFERENTES partitions.
Después del shuffle:
Partition 5:  {fare=10.5: [50, 25, 35]}  ← TODOS los 10.5 juntos
Partition 12: {fare=20.0: [30, 40]}      ← TODOS los 20.0 juntos
Partition 87: {fare=15.0: [20]}          ← TODOS los 15.0 juntos

CRÍTICO: El shuffle es la operación MÁS COST
•	Mueve datos por la red (si es cluster)
•	Escribe/lee del disco temporalmente
•	Es un "barrier" (todo debe completar antes de continuar)
