### Modelos CART (Classification and Regression Tree)

```
Los algoritmos basados en árboles se consideran uno de los mejores y más utilizados métodos de machine learning supervisado. Estos algoritmos dotan a los modelos predictivos de alta precisión, estabilidad y facilidad de interpretación. A diferencia de los modelos lineales, capturan bastante bien las relaciones no lineales. Son adaptables para resolver cualquier tipo de problema (clasificación o regresión).

Métodos como los árboles de decisión, random forest y gradient boosting se utilizan ampliamente en todo tipo de problemas de ciencia de datos. Por lo tanto, para todo analista (incluso principiantes), es importante aprender estos algoritmos y usarlos en la modelación.

Este notebook tiene como objetivo ayudar a los estudiantes a aprender los algoritmos basados en árboles desde cero. Al completar exitosamente este notebook, se espera que ustedes sean competentes en el uso de estos algoritmos y en la construcción de modelos predictivos.

Los árboles de decisión (DT) son un método de aprendizaje supervisado no paramétrico utilizado para la clasificación y la regresión.
El nodo superior de un árbol de decisión se conoce como nodo raíz, luego en cada bifurcación se divide según una variable explicativa o prdictora  y cada nodo final contiene una predicción para la variable objetivo. El objetivo es crear un modelo que prediga el valor de una variable objetivo mediante el aprendizaje de reglas de decisión simples inferidas a partir de las características de los datos. Un árbol puede considerarse una aproximación constante por partes.

No necesita StandardScaler. Los arboles no calculan distancias entre puntos, solo hacen preguntas binarias del tipo trip_distance > 3.5?. La escala de las variables es irrelevante para ese proceso. Parte 5 del codigo es exclusiva de CART. El toDebugString te imprime las reglas exactas en texto legible — algo que KNN y Logit no pueden hacer de esta forma. La importancia de variables ademas te dira cuanto aporta cada una de las tres al modelo, que es directamente comparable con los coeficientes beta del Logit.


Estructura de árbol: CART construye una estructura similar a un árbol compuesta por nodos y ramas.  En el árbol de decisión, los nodos se dividen en subnodos según un valor umbral de algún atributo. Cada bifurcación se divide según una variable predictora y cada nodo final contiene una predicción para la variable objetivo

El nodo raíz se toma del conjunto de entrenamiento y se divide en dos considerando el mejor atributo y su valor umbral. Luego, los subconjuntos también se dividen usando la misma lógica. Este proceso continúa hasta que se encuentra el último subconjunto puro en el árbol o hasta alcanzar el número máximo de hojas permitido durante el crecimiento del árbol.

El término CART se usa como un término genérico para las siguientes categorías de árboles de decisión:

- Árboles de Clasificación: El árbol se utiliza para determinar a qué “clase” es más probable que pertenezca la variable objetivo cuando esta es categórica.

- Árboles de Regresión: Se utilizan para predecir el valor de una variable continua.

Algunas ventajas de los arboles de decision:

Requiere poca preparación de datos. Otras técnicas suelen requerir la normalización de datos, la creación de variables ficticias y la eliminación de valores en blanco. Algunas combinaciones de árboles y algoritmos admiten valores faltantes.

Utiliza un modelo de caja blanca. Si una situación dada es observable en un modelo, la explicación de la condición se explica fácilmente mediante lógica booleana. Por el contrario, en un modelo de caja negra (por ejemplo, en una red neuronal artificial), los resultados pueden ser más difíciles de interpretar.

Los aprendices de árboles de decisión crean árboles sesgados si algunas clases predominan. Por lo tanto, se recomienda equilibrar el conjunto de datos antes de ajustarlo al árbol de decisión.

````python
from sklearn import tree
from sklearn.model_selection import GridSearchCV, StratifiedKFold, train_test_split
from sklearn.metrics import (classification_report, confusion_matrix, 
                              roc_auc_score, roc_curve)
import matplotlib.pyplot as plt
import pandas as pd

# ==========================================================
# 1. TRAIN/TEST SPLIT (antes del grid search)
# ==========================================================
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=23, stratify=y
)

# ==========================================================
# 2. INSTANCIAR EL CART
# ==========================================================
arbol = tree.DecisionTreeClassifier(random_state=23, class_weight='balanced')

# ==========================================================
# 3. GRILLA DE HIPERPARÁMETROS (completa)
# ==========================================================
grilla_arbol = {
    'criterion':         ['gini', 'entropy'],          # criterio de división
    'max_depth':         [3, 5, 7, 9, 11, 13, 15, None],  # profundidad máxima
    'min_samples_split': [2, 10, 20, 50, 100],         # mínimo para dividir un nodo
    'min_samples_leaf':  [1, 5, 10, 20, 50],           # mínimo en cada hoja
    'max_features':      [None, 'sqrt', 'log2'],       # nº de features por split
    'ccp_alpha':         [0.0, 0.001, 0.005, 0.01]     # post-pruning
}

# ==========================================================
# 4. GRID SEARCH
# ==========================================================
cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=23)

arbol_cv = GridSearchCV(
    estimator=arbol,
    param_grid=grilla_arbol,
    scoring='roc_auc',        # mejor que accuracy con clases desbalanceadas
    cv=cv,
    return_train_score=True,
    n_jobs=-1,                # usa todos los núcleos
    verbose=1,
    refit=True
)

arbol_cv.fit(X_train, y_train)

print(f"Mejores parámetros: {arbol_cv.best_params_}")
print(f"Mejor AUC (CV):     {arbol_cv.best_score_:.4f}")

# ==========================================================
# 5. EVALUACIÓN EN TEST
# ==========================================================
mejor_arbol = arbol_cv.best_estimator_
y_pred  = mejor_arbol.predict(X_test)
y_proba = mejor_arbol.predict_proba(X_test)[:, 1]

print("\n=== Matriz de confusión ===")
print(confusion_matrix(y_test, y_pred))

print("\n=== Reporte de clasificación ===")
print(classification_report(y_test, y_pred,
      target_names=['Ocupado', 'Desocupado']))

print(f"AUC en test: {roc_auc_score(y_test, y_proba):.4f}")

# Importancia de variables
importancias = pd.DataFrame({
    'variable':    X.columns,
    'importancia': mejor_arbol.feature_importances_
}).sort_values('importancia', ascending=False)
print("\n=== Importancia de variables ===")
print(importancias)
````

#### ¿Sobreajuste o subajuste?

Rendimiento en los datos de entrenamiento:

- El sobreajuste produce una precisión muy alta en el entrenamiento, mientras que el subajuste resulta en una baja precisión en el entrenamiento.

Rendimiento en los datos de prueba/validación:

- El sobreajuste causa un mal rendimiento en datos no vistos, mientras que el subajuste también tiene un mal desempeño en los datos de prueba o validación.

Profundidad baja → modelo simple → bajo accuracy en train y test (underfitting).
Profundidad muy alta → accuracy de train casi 1, pero test baja (overfitting).
En el medio → hay un punto óptimo (marcado por la línea roja).

#### Gini index/Gini impurity

El índice de Gini es una métrica para las tareas de clasificación en CART. Almacena la suma de las probabilidades al cuadrado de cada clase. Calcula el grado de probabilidad de que una variable específica sea clasificada erróneamente cuando se elige al azar, y es una variación del coeficiente de Gini. Funciona con variables categóricas, proporciona resultados de tipo “éxito” o “fracaso” y, por lo tanto, solo realiza divisiones binarias.

El valor del índice de Gini varía entre 0 y 1:

El índice de Gini mide qué tan “mezclados” están los datos en un nodo de un árbol de decisión.
- Si en un nodo todas las observaciones son de la misma clase, la impureza es 0 (perfectamente puro).
  
- Si en un nodo las clases están muy mezcladas, la impureza se acerca a 1 (muy impuro).

- Un valor de 1 – 1/n ocurre cuando los elementos están distribuidos uniformemente en n clases y cada clase tiene la misma probabilidad de 1/n. Por ejemplo, con dos clases, la impureza de Gini es 1-(1/2)=0,5 


$$Gini=1-\sum_{i = 1}^{n}p_{i}^{2}$$

En conclusión, la impureza de Gini es la probabilidad de clasificación errónea, asumiendo la selección independiente del elemento y de su clase en función de las probabilidades de clase

#### Ejemplo intuitivo: con bolas de colores

Imaginemonos  una caja con bolas rojas y azules:
1.  Caso puro (Gini = 0):
- Caja con 100% bolas rojas.

No hay confusión: todas son de una sola clase.

2. Caso muy mezclado (Gini ≈ 0.5 con 2 clases):
- Caja con 50% rojas y 50% azules.
- Si saco una bola al azar, tengo mucha incertidumbre (es casi igual de probable que sea roja o azul).

3. Caso intermedio (Gini entre 0 y 0.5):
- Caja con 80% rojas y 20% azules.
- Hay más certeza de que sacaré una roja, pero aún existe cierta mezcla.

$$
Gini= 1-(p^2_{rojo}+(p^2_{azul})
$$

si: 
$$
p_{rojo} = 1 \; Gini = 0
$$

si:
$$
p_{rojo} = 0.5 \; Gini = 0.5
$$

si: 

$$
p_{rojo} = 0.8 \; Gini = 1-(0.8^2-0.2^2)=0.32
$$

#### Elementos clave a interpretar


Elemento y 	Significado

- feature_name <= threshold.    Es una condición de división: si se cumple, vamos a la rama izquierda; si no, a la derecha.
- gini:  Es	una medida de impureza. Entre 0 (puro, un solo grupo) y 0.5 (máxima mezcla de clases en un nodo binario).
- samples:   Es la cantidad de muestras en ese nodo.
- value:  Es el	Recuento de muestras por clase. Ej: [90, 10] significa 90 clase 0, 10 clase 1.

### Métodos de Ensamblaje

Los métodos de ensamblado son técnicas poderosas en el aprendizaje automático que combinan múltiples modelos para mejorar la precisión general de las predicciones y la estabilidad del modelo. La agregación por remuestreo, mejor conocida como Bagging, se destaca como un método de ensamblado popular y ampliamente utilizado.

El modelado en conjunto (Ensemble Modeling) es una técnica que combina múltiples modelos de aprendizaje automático para mejorar el rendimiento predictivo general. La idea básica es que un grupo de modelos débiles puede unirse para formar un modelo fuerte. Un modelo en conjunto generalmente consta de dos pasos:

Se entrenan múltiples modelos de aprendizaje automático de forma independiente.
Sus predicciones se agregan de alguna manera, como por votación, promediado o ponderación. Este conjunto se utiliza luego para realizar la predicción final.

Los modelos en conjunto tienden a ofrecer mejores resultados porque los diferentes modelos se complementan entre sí y superan sus debilidades individuales. ## Bagging


Bagging (de Bootstrap Aggregating) es una técnica de ensamblado en ciencia de datos y aprendizaje automático que mejora la precisión y estabilidad de modelos débiles (como árboles de decisión) mediante la combinación de múltiples modelos entrenados sobre subconjuntos aleatorios del conjunto de datos original.

#### Cómo funciona Bagging?

- Se generan múltiples subconjuntos del conjunto de entrenamiento original mediante muestreo con reemplazo (bootstrap).
  
- Se entrena un modelo independiente en cada subconjunto (por ejemplo, un árbol de decisión).

- Para clasificación, se combinan los resultados por votación mayoritaria; para regresión, por promedio.


#### Ventajas
- Reduce la varianza del modelo.
  
- Mitiga el sobreajuste.

- Funciona bien con modelos inestables (como árboles de decisión).


````python
# reading png image file 
# importing pyplot and image from matplotlib 
import matplotlib.pyplot as plt 
import matplotlib.image as img 
im = img.imread('Captura de pantalla 2025-03-08 a la(s) 1.27.05 .png') 
  # show image 
plt.figure(figsize=(12, 12))  # Cambia los valores para ajustar el tamaño deseado
plt.axis('off')
plt.imshow(im)
plt.show()
````

En detalle, cada modelo se entrena con un subconjunto aleatorio de los datos muestreado con reemplazo, lo que significa que los mismos puntos de datos pueden ser seleccionados más de una vez. Este subconjunto aleatorio se conoce como una muestra bootstrap. Al entrenar modelos en diferentes muestras bootstrap, el bagging reduce la varianza de los modelos individuales. También evita el sobreajuste al exponer los modelos componentes a diferentes partes del conjunto de datos. Las predicciones de todos los modelos muestreados se combinan luego mediante un simple promedio para generar la predicción final. De esta manera, el modelo agregado incorpora las fortalezas de los modelos individuales y cancela sus errores. El bagging es particularmente eficaz para reducir la varianza y el sobreajuste, haciendo que el modelo sea más robusto y preciso, especialmente en casos donde los modelos individuales son propensos a una alta variabilidad.

````python
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
````

### Otra manera de hacerlo con Pipeline

Pipeline es una herramienta para encadenar pasos de preprocesamiento y modelado.

````python
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('classifier', DecisionTreeClassifier(random_state=42))
])

pipeline.fit(X_train, y_train)

#'scaler': aplica StandardScaler a las variables de entrada X.
#'classifier': entrena un DecisionTreeClassifier sobre los datos ya transformados.
````

Ese código crea un flujo de trabajo donde primero se estandarizan los datos (StandardScaler) y después se entrena un árbol de decisión (DecisionTreeClassifier) con ellos. Aunque en este caso la estandarización no es necesaria, la ventaja de usar Pipeline es que luego puedes aplicar los mismos pasos de preprocesamiento de manera consistente en validación cruzada o en datos de prueba.

````python
# Predecir
y_pred = pipeline.predict(X_test)

# Evaluar
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy con pipeline: {accuracy * 100:.2f}%")
````

#### Buenas prácticas y recomendaciones

Al utilizar bagging en aprendizaje automático, seguir buenas prácticas y consejos puede maximizar su efectividad:

- Utiliza bagging cuando trabajes con modelos de alta varianza, como los árboles de decisión, para mejorar su estabilidad.

- Combina bagging con validación cruzada para una evaluación más confiable de tus modelos.

- Usa random forests (árboles de decisión con bagging) como una técnica poderosa y lista para usar.

- Establece un valor alto para n_estimators, como entre 100 y 200, para obtener el máximo beneficio.

- El bagging se puede paralelizar fácilmente usando n_jobs. Implementa el entrenamiento en múltiples CPU o máquinas para mayor rapidez.

- Dado que el bagging se basa en muestreo bootstrap, asegúrate de que cada modelo se entrene con subconjuntos suficientemente diversos de los datos.

- Antes de agregar modelos, optimiza el rendimiento de cada uno con GridSearchCV. Un buen rendimiento individual suele traducirse en un mejor rendimiento del conjunto.

### Bagging vs. Boosting

Boosting es otro método de conjunto popular que a menudo se compara con bagging. La principal diferencia radica en cómo se entrenan los modelos que lo componen.

En bagging, los modelos se entrenan de forma independiente y en paralelo sobre distintos subconjuntos aleatorios de los datos. En cambio, en boosting, los modelos se entrenan de manera secuencial, donde cada modelo aprende de los errores del anterior. Además, bagging suele combinar los modelos mediante un promedio simple, mientras que boosting asigna pesos a los modelos según su precisión.

````python
# reading png image file 
# importing pyplot and image from matplotlib 
import matplotlib.pyplot as plt 
import matplotlib.image as img 
im = img.imread('Captura de pantalla 2025-03-08 a la(s) 1.41.39 .png') 
  # show image 
plt.figure(figsize=(12, 12))  # Cambia los valores para ajustar el tamaño deseado
plt.axis('off')
plt.imshow(im)
plt.show()
````

Bagging reduce la varianza, mientras que boosting reduce el sesgo. Bagging puede utilizarse con modelos inestables como los árboles de decisión, mientras que boosting funciona mejor con modelos estables como la regresión lineal.

Ambos métodos tienen sus fortalezas y debilidades. Bagging es más sencillo de ejecutar en paralelo, mientras que boosting puede ser más potente y preciso. En la práctica, es útil probar ambos enfoques en un nuevo problema para ver cuál ofrece un mejor rendimiento.

#### Diferentes tipos de métodos en conjunto

- Boosting es una técnica destacada de aprendizaje en conjunto, pero es solo una entre varias que mejoran la capacidad predictiva de los modelos. Veamos algunas otras:

- Bagging (Bootstrap Aggregating): Método que entrena múltiples modelos sobre subconjuntos aleatorios de los datos de entrenamiento y agrega sus predicciones. Reduce la varianza y ayuda a evitar el sobreajuste.

- Stacking (Stacked Generalization): Técnica que combina múltiples modelos entrenando un meta-modelo que aprende cómo combinar mejor las predicciones de los modelos base. Puede capturar patrones complejos que los modelos individuales podrían pasar por alto.

- Blending: Similar al stacking, pero utiliza un conjunto de validación separado para entrenar el meta-modelo en lugar de validación cruzada. Es más simple y rápido que el stacking, aunque puede ser menos robusto.

- Voting: Combina las predicciones de múltiples modelos mediante voto mayoritario (votación dura) o promedio ponderado de probabilidades predichas (votación blanda). Es sencillo de implementar y puede ser eficaz cuando se utilizan modelos base diversos.

## Boosting 

Boosting mejora el rendimiento del aprendizaje automático corrigiendo errores de forma secuencial y combinando modelos débiles en predictores fuertes.

Los métodos en conjunto (Ensemble methods) son técnicas poderosas en el aprendizaje automático que combinan múltiples modelos para mejorar la precisión de las predicciones y la estabilidad del modelo. El Bootstrap Aggregating, más conocido como Bagging, se destaca como un método en conjunto popular y ampliamente implementado.

Los avances recientes en aprendizaje automático han introducido nuevas formas de abordar problemas complejos. Boosting es una técnica que sigue mostrando gran potencial. Está cambiando la manera en que abordamos el modelado de datos al utilizar múltiples algoritmos para mejorar el rendimiento. A medida que el concepto de boosting continúa evolucionando, han surgido nuevas variantes como Gradient Boosting y XGBoost, que están ampliando los límites de la velocidad y la precisión.

### Cómo funciona el Boosting

Boosting transforma modelos débiles en un único modelo fuerte mediante un proceso sistemático que se enfoca en reducir errores a través del entrenamiento secuencial de modelos. Los pasos involucrados incluyen:

Seleccionar pesos iniciales: Se asignan pesos iniciales a todos los puntos de datos para indicar su importancia en el proceso de aprendizaje.
Entrenar secuencialmente: Se entrena el primer modelo débil con los datos. Tras evaluar su desempeño, se incrementan los pesos de las instancias mal clasificadas. Esto hace que el siguiente modelo débil se enfoque más en los casos difíciles.
Repetir el proceso: Se repite el ajuste de pesos y el entrenamiento de nuevos modelos. Cada nuevo modelo se centra en las debilidades del conjunto construido hasta ese momento.
Combinar los resultados: Se agregan las predicciones de todos los modelos débiles para formar la salida final. Esta agregación suele ser ponderada, dando mayor influencia a los modelos más precisos.
Este método minimiza eficazmente los errores al enfocarse de manera más intensa en los casos difíciles dentro de los datos de entrenamiento, lo que da como resultado un rendimiento predictivo sólido.

### Tipos de algoritmos de Boosting

Veamos algunos de los algoritmos de boosting más conocidos.

#### -  AdaBoost (Adaptive Boosting)

AdaBoost es uno de los primeros algoritmos de boosting. Se enfoca en reasignar pesos a los ejemplos de entrenamiento cada vez que se añade un nuevo modelo, dando más importancia a las instancias clasificadas incorrectamente. AdaBoost es especialmente eficaz para problemas de clasificación binaria. Consulta nuestro tutorial sobre el clasificador AdaBoost en Python para aprender más.

#### - Gradient Boosting

Gradient Boosting construye modelos de forma secuencial y corrige los errores a lo largo del proceso. Utiliza un algoritmo de descenso por gradiente para minimizar la función de pérdida al añadir nuevos modelos. Este método es flexible y puede aplicarse tanto a problemas de regresión como de clasificación.

#### - XGBoost (Extreme Gradient Boosting)

XGBoost es una biblioteca optimizada de gradient boosting distribuido y la opción preferida de muchos ganadores de competiciones en Kaggle. Está diseñada para ser altamente eficiente, flexible y portátil. Implementa algoritmos de aprendizaje automático bajo el marco de Gradient Boosting, ofreciendo una solución escalable y precisa para muchos problemas prácticos con datos. Para un estudio más detallado, te recomendamos revisar nuestro tutorial Usando XGBoost en Python y tomar nuestro curso especializado: Extreme Gradient Boosting with XGBoost.

### Random Forest

Random Forest es un algoritmo de aprendizaje supervisado. Tiene dos variantes: una se utiliza para problemas de clasificación y la otra para problemas de regresión. Es uno de los algoritmos más flexibles y fáciles de usar. Crea árboles de decisión a partir de muestras de datos, obtiene una predicción de cada árbol y selecciona la mejor solución mediante votación. Además, es un muy buen indicador de la importancia de las variables.

El algoritmo de Random Forest combina múltiples árboles de decisión, lo que da como resultado un "bosque" de árboles, de ahí el nombre Random Forest (bosque aleatorio). En el clasificador Random Forest, mientras mayor sea el número de árboles en el bosque, mayor será la precisión del modelo.

Intuición del algoritmo Random Forest
La intuición detrás del algoritmo Random Forest puede dividirse en dos etapas.

Primera etapa: construcción del bosque

En esta etapa, se seleccionan aleatoriamente "k" características de un total de "m" características para construir el bosque aleatorio. El proceso es el siguiente:

- Seleccionar aleatoriamente k características de un total de m características, donde k < m.
- Entre esas k características, calcular el nodo d utilizando el mejor punto de división (split).
- Dividir el nodo en nodos hijos usando ese mejor punto de división.
- Repetir los pasos 1 a 3 hasta alcanzar l nodos (según la profundidad deseada o criterios de parada).
- Construir el bosque repitiendo los pasos 1 a 4 n veces, para crear n árboles de decisión.

El algoritmo Random Forest puede usarse en el proceso de selección de características. Este algoritmo permite clasificar o ordenar la importancia de las variables en un problema de regresión o clasificación.

Se mide la importancia de una variable en un conjunto de datos ajustando el algoritmo de Random Forest a los datos. Durante el proceso de ajuste, se registra el error fuera de bolsa (out-of-bag error) para cada punto de datos y se promedia sobre todos los árboles del bosque.

La importancia de la j-ésima variable se mide luego del entrenamiento. Los valores de dicha variable se reordenan aleatoriamente entre los datos de entrenamiento, y se vuelve a calcular el error fuera de bolsa con este conjunto de datos perturbado. La puntuación de importancia para la j-ésima variable se obtiene promediando la diferencia en el error fuera de bolsa antes y después de la permutación, sobre todos los árboles. Esta puntuación se normaliza usando la desviación estándar de esas diferencias.

Las características que producen valores altos en esta puntuación son consideradas más importantes que aquellas con puntuaciones bajas. Con base en esta puntuación, se eligen las características más relevantes y se eliminan las menos importantes para construir el modelo.

Random Forest, o Bosque de Decisión Aleatorio, es un algoritmo supervisado de aprendizaje automático que se utiliza para clasificación, regresión y otras tareas, usando árboles de decisión.

Los Random Forests son especialmente adecuados para trabajar con conjuntos de datos grandes y complejos, manejar espacios de características de alta dimensión, y proporcionar información sobre la importancia de las variables. La capacidad de este algoritmo para mantener una alta precisión predictiva mientras minimiza el sobreajuste lo convierte en una opción popular en diversos campos, como las finanzas, la salud y el análisis de imágenes, entre otros.

El clasificador Random Forest crea un conjunto de árboles de decisión a partir de un subconjunto aleatorio del conjunto de entrenamiento. Luego, recolecta los votos de los distintos árboles para decidir la predicción final.

#### Diferencia entre Random Forest y Árboles de Decisión

Voy a comparar los Random Forests con los árboles de decisión. Algunas características destacadas de la comparación son las siguientes:

- Random Forest es un conjunto de múltiples árboles de decisión.

- Los árboles de decisión son computacionalmente más rápidos en comparación con los Random Forests.

- Los árboles de decisión muy profundos pueden sufrir de sobreajuste. El Random Forest evita el sobreajuste creando árboles sobre subconjuntos aleatorios del conjunto de datos.

- El Random Forest es difícil de interpretar, mientras que un árbol de decisión es fácilmente interpretable y puede convertirse en un conjunto de reglas.

Una relación entre los Random Forests y el algoritmo de k vecinos más cercanos (k-NN) fue señalada por Lin y Jeon en 2002. Resulta que ambos pueden considerarse como esquemas de vecindarios ponderados (weighted neighbourhoods schemes). Estos son modelos construidos a partir de un conjunto de entrenamiento que hacen predicciones para nuevos puntos observando el vecindario del punto, formalizado mediante una función de ponderación.
