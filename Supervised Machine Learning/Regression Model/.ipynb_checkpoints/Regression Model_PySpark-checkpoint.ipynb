````python
# ============================================================
# REGRESIÓN LINEAL - NYC Yellow Taxi 2024
# Enfoque: Inferencia estadística completa
# ============================================================


from pyspark.sql import SparkSession
from pyspark.sql.functions import col, hour, dayofweek, month
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.regression import LinearRegression
from pyspark.ml.evaluation import RegressionEvaluator



# ============================================================
# 1. CARGAR Y LIMPIAR
# ============================================================
df = spark.read.parquet("yellow_tripdata_2024-01.parquet")

df_clean = df \
    .filter(col("total_amount").between(1, 500)) \
    .filter(col("trip_distance").between(0.1, 100)) \
    .filter(col("passenger_count").between(1, 6)) \
    .filter(col("fare_amount") > 0) \
    .dropna(subset=["total_amount", "trip_distance", "passenger_count",
                    "fare_amount", "tip_amount", "tolls_amount"]) \

# ============================================================
# 2. FEATURES Y TARGET
# ============================================================
feature_cols = [
    "trip_distance",
    "passenger_count",
    "fare_amount",
    "tip_amount",
    "tolls_amount"
    
]

assembler = VectorAssembler(inputCols=feature_cols, outputCol="features")
df_model  = assembler.transform(df_clean.select(feature_cols + ["total_amount"]))

train_df, test_df = df_model.randomSplit([0.8, 0.2], seed=42)
````

#### Estimacion del modelo

$$
\widehat{Y} = \hat{\beta}_0 + \hat{\beta}_1 X_1 + \hat{\beta}_2 X_2 + \hat{\beta}_3 X_3 + \hat{\beta}_4 X_4 + \hat{\beta}_5 X_5
$$

$$\widehat{\text{total\_amount}} = \hat{\beta}_0 + \hat{\beta}_1(\text{trip\_distance}) + \hat{\beta}_2(\text{passenger\_count}) + \hat{\beta}_3(\text{fare\_amount}) + \hat{\beta}_4(\text{tip\_amount}) + \hat{\beta}_5(\text{tolls\_amount})
$$


$$\widehat{\text{total\_amount}} = 4.7987 + 0.1264 \cdot \text{trip\_distance} + 0.0032 \cdot \text{passenger\_count} + 0.9607 \cdot \text{fare\_amount} + 1.0979 \cdot \text{tip\_amount} + 1.2093 \cdot \text{tolls\_amount}
$$


````python
# ============================================================
# 3. REGRESIÓN LINEAL
# ============================================================
lr = LinearRegression(
    featuresCol="features",
    labelCol="total_amount",
    maxIter=100,
    regParam=0.0,        # Sin regularización → inferencia clásica
    elasticNetParam=0.0,
    solver="normal"      # Solución exacta (mínimos cuadrados)
)

modelo = lr.fit(train_df)
summary = modelo.summary   # Objeto con todas las métricas

# ============================================================
# 4. COEFICIENTES E INTERCEPTO
# ============================================================
print("=" * 65)
print("       COEFICIENTES ESTIMADOS")
print("=" * 65)
print(f"{'Variable':<20} {'Coef':>10} {'Std Error':>12} {'t-stat':>10} {'p-value':>10}")
print("-" * 65)

coefs     = modelo.coefficients
std_errs  = summary.coefficientStandardErrors   # incluye intercepto al final
t_values  = summary.tValues
p_values  = summary.pValues

# Los últimos valores corresponden al intercepto
for i, nombre in enumerate(feature_cols):
    sig = ""
    if p_values[i] < 0.001: sig = "***"
    elif p_values[i] < 0.01: sig = "**"
    elif p_values[i] < 0.05: sig = "*"
    elif p_values[i] < 0.1:  sig = "."

    print(f"{nombre:<20} {coefs[i]:>10.4f} {std_errs[i]:>12.4f} "
          f"{t_values[i]:>10.4f} {p_values[i]:>10.4f} {sig}")

# Intercepto (último elemento)
sig_int = "***" if p_values[-1] < 0.001 else ""
print(f"{'(Intercepto)':<20} {modelo.intercept:>10.4f} {std_errs[-1]:>12.4f} "
      f"{t_values[-1]:>10.4f} {p_values[-1]:>10.4f} {sig_int}")

print("-" * 65)
print("Signif.: *** p<0.001  ** p<0.01  * p<0.05  . p<0.1")
````

#### Interpretacion.


```
- Signos de los estimadores:

Todos los signos de los estimadores son positivos, esto quiere decir que la relacion entre cada variable explicativa y la variable dependiente es directa (comparten la misma dirección, si la variable Xi aumenta, Y aumenta, si la variable Xi disminuye, Y disminuye

- Inferencia estadistica individual:

Dado que el nivel de probabilidad es menor que el nivel de significancia escogido (5% o 0.05), todas al variables son significativas individualmente.


#### Metricas de ajuste y de pronostico del modelo



````python
# ============================================================
# 5. MÉTRICAS DE AJUSTE DEL MODELO
# ============================================================
pred_test = modelo.transform(test_df)

evaluator = RegressionEvaluator(labelCol="total_amount", predictionCol="prediction")

rmse = evaluator.setMetricName("rmse").evaluate(pred_test)
mae  = evaluator.setMetricName("mae").evaluate(pred_test)
r2   = evaluator.setMetricName("r2").evaluate(pred_test)

n    = train_df.count()
k    = len(feature_cols)
r2_adj = 1 - (1 - summary.r2) * (n - 1) / (n - k - 1)

print("\n")
print("=" * 40)
print("       MÉTRICAS DE AJUSTE")
print("=" * 40)
print(f"  R²  (train):          {summary.r2:.6f}")
print(f"  R²  ajustado (train): {r2_adj:.6f}")
print(f"  RMSE (train):         {summary.rootMeanSquaredError:.4f}")
print(f"  RMSE (test):          {rmse:.4f}")
print(f"  MAE  (test):          {mae:.4f}")
print(f"  R²   (test):          {r2:.6f}")
print(f"  Observaciones:        {n:,}")
print(f"  Variables:            {k}")
print("=" * 40)
````

#### Interpretacion metricas^^
```
- El coeficiente R² que se obtiene solo para train, nos dice que el modelo explica a la variable dependiente en un 99.49%
- El RMSE nos dice que el pronostico se equivoca en promedio 1.54 dolares

````python
# ============================================================
# 7. MUESTRA DE PREDICCIONES VS REAL
# ============================================================
import pyspark.sql.functions as F

print("\n>>> Predicciones vs Valor Real (10 ejemplos):")
pred_test.select(
    "trip_distance",
    "fare_amount",
    "total_amount",
    "prediction",
    F.round(col("total_amount") - col("prediction"), 2).alias("residual")
).show(10, truncate=False)
````

#### Inferencia global

````python
# ============================================================
#  INFERENCIA GLOBAL — TEST F
# Hipótesis: H0: todos los coeficientes = 0
#            H1: al menos uno ≠ 0
# ============================================================
from scipy import stats as scipy_stats

# SCE = Suma de Cuadrados Explicada
# SCR = Suma de Cuadrados de Residuos
# SCT = Suma de Cuadrados Total

sce   = summary.r2 * summary.rootMeanSquaredError**2 * n / (1 - summary.r2) \
        if summary.r2 < 1 else float("inf")

scr   = summary.rootMeanSquaredError**2 * n
sct   = scr + sce

# Estadístico F
f_stat = (sce / k) / (scr / (n - k - 1))
p_fstat = scipy_stats.f.sf(f_stat, dfn=k, dfd=n - k - 1)

print("\n")
print("=" * 55)
print("       TABLA 3 — INFERENCIA GLOBAL (Test F)")
print("=" * 55)
print(f"  H0: todos los coeficientes son simultáneamente = 0")
print(f"  H1: al menos un coeficiente ≠ 0")
print("-" * 55)
print(f"  SCE (explicada):      {sce:>15.4f}  gl = {k}")
print(f"  SCR (residuos):       {scr:>15.4f}  gl = {n-k-1:,}")
print(f"  SCT (total):          {sct:>15.4f}  gl = {n-1:,}")
print("-" * 55)
print(f"  CME (SCE/k):          {sce/k:>15.4f}")
print(f"  CMR (SCR/n-k-1):      {scr/(n-k-1):>15.4f}")
print("-" * 55)
print(f"  Estadístico F:        {f_stat:>15.4f}")
print(f"  p-value:              {p_fstat:>15.2e}")
print(f"  Resultado:            {'Rechazar H0 ***' if p_fstat < 0.05 else 'No rechazar H0'}")
print("=" * 55)
print("\nInterpretación:")
print("  Si p-value < 0.05 → el modelo es globalmente significativo")
print("  Al menos una variable explica la variación en total_amount")
````
